```python
import requests
import time
import arxiv
import os
from tqdm import tqdm
import pandas as pd

def get_arxiv_pdf(arxiv_id):
    try:
        search = arxiv.Search(id_list=[arxiv_id])
        paper = next(search.results())
        return paper.pdf_url
    except Exception as e:
        print(f"\nError getting arXiv PDF URL for {arxiv_id}: {e}")
        return None

def retrieve_url(url, filepath):
    try:
        response = requests.get(url, stream=True)
        response.raise_for_status()
        with open(filepath, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        return True
    except Exception as e:
        print(f"\nError downloading {url}: {e}")
        return False

def get_all_citations(paper_id):
    citations = []
    offset = 0
    limit = 1000
    total_citations = 0
    
    url = f"http://api.semanticscholar.org/graph/v1/paper/{paper_id}"
    params = {"fields": "citationCount"}
    try:
        response = requests.get(url, params=params)
        if response.status_code == 200:
            total_citations = response.json()['citationCount']
            print(f"Total citations to fetch: {total_citations}")
        else:
            print(f"Failed to get citation count: {response.status_code}")
            return citations
    except Exception as e:
        print(f"Error getting citation count: {e}")
        return citations
    
    with tqdm(total=total_citations, desc="Fetching citations") as pbar:
        while True:
            url = f"http://api.semanticscholar.org/graph/v1/paper/{paper_id}/citations"
            params = {
                "fields": "citingPaper.title,citingPaper.externalIds,citingPaper.url",
                "limit": limit,
                "offset": offset
            }
            
            try:
                response = requests.get(url, params=params)
                if response.status_code == 200:
                    data = response.json()
                    batch = data.get('data', [])
                    if not batch:
                        break
                    
                    citations.extend(batch)
                    pbar.update(len(batch))
                    
                    offset += limit
                    time.sleep(1)
                else:
                    print(f"Error fetching citations: {response.status_code}")
                    break
            except Exception as e:
                print(f"Error in citation request: {e}")
                break
                
    return citations

paperId = "1b6e810ce0afd0dd093f789d2b2742d047e316d5"
citations = get_all_citations(paperId)

arxiv_citations = []
for citation in citations:
    if 'citingPaper' in citation:
        paper = citation['citingPaper']
        external_ids = paper.get('externalIds', {})
        if any(key.lower() == 'arxiv' for key in external_ids.keys()):
            arxiv_citations.append(paper)

print(f"\nTotal citations with arXiv IDs: {len(arxiv_citations)}")
print("\nFirst 5 papers to be downloaded:")
for i, paper in enumerate(arxiv_citations[:5]):
    print(f"{i+1}. {paper.get('title')}")

print("\nStarting downloads...")
if not os.path.exists("pdfs"):
    os.makedirs("pdfs")

for paper in tqdm(arxiv_citations, desc="Processing arXiv papers"):
    title = paper['title']
    arxiv_key = next(key for key in paper['externalIds'].keys() if key.lower() == 'arxiv')
    arxiv_id = paper['externalIds'][arxiv_key]
    
    safe_title = "".join(c for c in title if c.isalnum() or c in (' ', '-', '_')).strip()
    pdf_path = f"pdfs/{safe_title}.pdf"
    
    if os.path.exists(pdf_path):
        continue
        
    pdf_url = get_arxiv_pdf(arxiv_id)
    if pdf_url:
        if retrieve_url(pdf_url, pdf_path):
            continue  
        else:
            print(f"\nFailed to download: {title}")
    
    time.sleep(3)
```

```python
from pathlib import Path
from PyPDF2 import PdfReader
import json
from tqdm import tqdm
import warnings
import logging

logger = logging.getLogger("PyPDF2")
logger.setLevel(logging.ERROR)

warnings.filterwarnings('ignore', message='.*FloatObject.*invalid.*')
warnings.filterwarnings('ignore', message='.*unknown widths.*')
warnings.filterwarnings('ignore', category=Warning, module='PyPDF2')

def clean_text(text):
    """Clean text of invalid Unicode characters"""
    if not text:
        return ""
    # Replace surrogate characters with a replacement character
    return text.encode('utf-16', 'surrogatepass').decode('utf-16', 'replace')

directory = "pdfs"
pdf_files = list(Path(directory).glob('*.pdf'))
pdf_text_dict = {}

# Create progress bar
for file in tqdm(pdf_files, desc="Processing PDFs"):
    try:
        reader = PdfReader(file)
        meta = reader.metadata
        
        # Initialize dictionary entry with metadata, cleaning each field
        pdf_text_dict[file.stem] = {
            "filename": file.name,
            "author": clean_text(meta.author) if meta.author else "",
            "creator": clean_text(meta.creator) if meta.creator else "",
            "subject": clean_text(meta.subject) if meta.subject else "",
            "title": clean_text(meta.title) if meta.title else "",
            "text": ""
        }
        
        # Extract text from all pages
        text_content = []
        for page in reader.pages:
            try:
                extracted_text = page.extract_text()
                if extracted_text:
                    text_content.append(clean_text(extracted_text))
            except Exception as page_error:
                print(f"\nError extracting text from page in {file}: {str(page_error)}")
                continue
            
        # Join all pages' text with newlines
        pdf_text_dict[file.stem]["text"] = "\n".join(text_content)
        
    except Exception as e:
        print(f"\nError processing {file}: {str(e)}")
        pdf_text_dict[file.stem] = {
            "filename": file.name,
            "error": str(e),
            "text": ""
        }

# Save to JSON file
output_path = "pdf_contents.json"
try:
    # First try with standard encoding
    try:
        with open(output_path, 'w', encoding='utf-8') as f:
            json.dump(pdf_text_dict, f, ensure_ascii=False, indent=4)
    except UnicodeEncodeError:
        # If that fails, fall back to ensuring ASCII with escaping
        with open(output_path, 'w', encoding='utf-8') as f:
            json.dump(pdf_text_dict, f, ensure_ascii=True, indent=4)
    print(f"\nSuccessfully saved data to {output_path}")
except Exception as e:
    print(f"\nError saving JSON file: {str(e)}")
    # Last resort: try to save with problematic characters removed
    try:
        cleaned_dict = {k: {
            key: str(value).encode('ascii', 'ignore').decode('ascii') 
            for key, value in v.items()
        } for k, v in pdf_text_dict.items()}
        with open(output_path, 'w', encoding='utf-8') as f:
            json.dump(cleaned_dict, f, ensure_ascii=True, indent=4)
        print(f"Saved file with ASCII-only characters to {output_path}")
    except Exception as final_e:
        print(f"Final attempt to save failed: {str(final_e)}")
```

```python
import json
import arxiv
import time
import re
from datetime import datetime
from tqdm import tqdm

def clean_title(title):
    """Clean and normalize the title for better matching"""
    # Remove special characters and extra whitespace
    cleaned = re.sub(r'[^\w\s-]', '', title)
    cleaned = ' '.join(cleaned.split())
    return cleaned

def fetch_arxiv_date(title):
    """
    Fetch publication date from arXiv given a paper title.
    Try multiple search strategies.
    """
    client = arxiv.Client()
    
    # Strategy 1: Exact title match
    search = arxiv.Search(
        query=f'ti:"{title}"',
        max_results=1
    )
    result = next(client.results(search), None)
    if result:
        return result.published.strftime('%Y-%m-%d'), result.entry_id

    time.sleep(1)  # Rate limiting between attempts
    
    # Strategy 2: Search without quotes and ti: prefix
    cleaned_title = clean_title(title)
    search = arxiv.Search(
        query=cleaned_title,
        max_results=5  # Get more results to find potential matches
    )
    
    # Check each result for title similarity
    results = list(client.results(search))
    for result in results:
        result_title = clean_title(result.title)
        # Check if titles are very similar
        if (cleaned_title.lower() in result_title.lower() or 
            result_title.lower() in cleaned_title.lower()):
            return result.published.strftime('%Y-%m-%d'), result.entry_id
    
    time.sleep(1)  # Rate limiting between attempts
    
    # Strategy 3: Try with first few words of title
    first_words = ' '.join(cleaned_title.split()[:4])  # First 4 words
    search = arxiv.Search(
        query=f'ti:"{first_words}"',
        max_results=5
    )
    
    results = list(client.results(search))
    for result in results:
        result_title = clean_title(result.title)
        if cleaned_title.lower() in result_title.lower():
            return result.published.strftime('%Y-%m-%d'), result.entry_id
            
    return None, None

# Load your JSON data
with open('papers_analysis_iterfinal_20250107_200800.json', 'r') as f:
    papers_data = json.load(f)

# Add dates to the data
for title in tqdm(papers_data.keys(), desc="Fetching paper dates", unit="paper"):
    paper_info = papers_data[title]
    date, arxiv_id = fetch_arxiv_date(title)
    if date:
        paper_info['publication_date'] = date
        paper_info['arxiv_id'] = arxiv_id
        # tqdm.write(f"✓ Found date for: {title} ({date})")
    else:
        continue
        # tqdm.write(f"✗ Could not find date for: {title}")
    # Respect rate limits
    time.sleep(2)

# Save the updated data
with open('papers_data_with_dates.json', 'w') as f:
    json.dump(papers_data, f, indent=4)

print("\nDone! Checking results...")
# Print summary of papers without dates
missing_dates = [title for title, info in papers_data.items() 
                if 'publication_date' not in info]
if missing_dates:
    print(f"\nCould not find dates for {len(missing_dates)} papers:")
    for title in missing_dates:
        print(f"- {title}")
else:
    print("\nFound dates for all papers!")
```

```python
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from datetime import datetime
from collections import Counter
import re
import numpy as np
from matplotlib.dates import DateFormatter
import matplotlib.dates as mdates
from scipy import stats

model_display_names = {
    'gpt-4': 'GPT-4',
    'gpt-3.5': 'GPT-3.5-turbo',
    'gpt-3': 'GPT-3',
    'palm': 'PaLM-2',
    'gpt-4o': 'GPT-4o',
    'llama-2-7b': 'LLaMA-2-7B',
    'bert': 'BERT',
    'llama-8b': 'LLaMA-3-8B',
    'mistral-7b': 'Mistral 7B',
    'llama': 'LLaMA',
    'llama-7b': 'LLaMA-7B',
    'llama-2-13b': 'LLAMA-2-13B',
    'gpt-4v': 'GPT-4V',
    'llama-70b': 'LLaMA-3-70B',
    'llama-2-70b': 'LLaMA-2-70B',
    't5': 'T5',
    'roberta': 'RoBERTa',
    'llama-2': 'LLaMA-2',
    'gpt-2': 'GPT-2',
    'llama-2-chat-7b': 'LLaMA-2-7B-Chat'
}

benchmark_display_names = {
    'mmlu': 'MMLU',
    'truthfulqa': 'TruthfulQA',
    'hellaswag': 'HellaSwag',
    'arc': 'ARC',
    'gsm8k': 'GSM8K',
    'boolq': 'BoolQ',
    'lambada': 'LAMBADA',
    'triviaqa': 'TriviaQA',
    'squad': 'SQuAD',
    'winogrande': 'Winogrande',
    'bbh': 'BBH',
    'math': 'MATH',
    'fever': 'FEVER',
    'openbookqa': 'OpenBookQA',
    'drop': 'DROP',
    'bigbench': 'BIG-bench',
    'glue': 'GLUE',
    'superglue': 'SuperGLUE',
    'natural-questions': 'Natural Questions',
    'c-eval': 'C-Eval',
    'math-qa': 'MathQA',
    'hotpotqa': 'HotpotQA',
    'aqua': 'AQuA',
    'piqa': 'PIQA',
    'hendrycks-test': 'MMLU',
    'nq': 'Natural Questions',
    'commonsenseqa': 'CommonsenseQA',
    'arc-challenge': 'ARC-Challenge',
    'arc-easy': 'ARC-Easy'
}

def normalize_model_name(name: str) -> tuple:
    """
    Normalizes model names while preserving important distinctions between model sizes
    and architectural variants. Returns tuple of (normalized_name, original_name)
    """
    original = name.strip('"').strip()
    normalized = original.lower()
    normalized = re.sub(r'\s+', ' ', normalized)
    
    # Extract parameter count/size information
    size_pattern = r'[\(\[\s]?(\d+[bm])[\)\]\s]?'
    size_match = re.search(size_pattern, normalized, re.IGNORECASE)
    model_size = size_match.group(1).lower() if size_match else None
    
    # First check for GPT-4 variants
    gpt4_patterns = [
        'gpt4', 'gpt 4', 'gpt-4', 'gpt-4-turbo', 'gpt4-turbo',
        'chatgpt-4', 'chat-gpt-4', 'chatgpt-4.0', 'chat gpt 4', 
        'chatgpt 4', 'gpt-4 turbo', 'gpt 4 turbo'
    ]
    
    if any(re.match(f'^{re.escape(pattern)}', normalized) for pattern in gpt4_patterns):
        if any(v in normalized for v in ['vision', '-v', 'v']):
            return 'gpt-4v', original
        elif any(o in normalized for o in ['4o', '4-o', '4 o']):
            return 'gpt-4o', original
        return 'gpt-4', 'GPT-4'
    
    # Model family mappings with their variants
    base_models = {
        'gpt-3.5': [
            'gpt-3.5-turbo', 'gpt3.5-turbo', 'gpt3.5 turbo', 
            'gpt-3.5-t', 'chatgpt', 'chat-gpt', 'chat gpt', 
            'chatgpt-3.5', 'chatgpt-3.5-turbo', 'chat-gpt-3.5',
            'chat gpt 3.5', 'chatgpt 3.5', 'gpt3.5'
        ],
        'gpt-3': [
            'gpt-3', 'gpt3', 'gpt 3', 'gpt-3-davinci',
            'davinci', 'text-davinci', 'davinci-002',
            'chatgpt-3', 'chat-gpt-3', 'gpt-3-(175b)',
            'gpt-3 (175b)', 'gpt-3 (davinci)'
        ],
        'llama-2': [
            'llama2', 'llama 2', 'llama-2', 'llama2-', 
            'llama-2-', 'llama 2 ', 'llama2 ', 'llama-2 '
        ],
        'llama': [
            'llama1', 'llama 1', 'llama-1', 'llama-', 
            'llama ', 'llamab', 'llama-b'
        ],
        'mistral': [
            'mistral', 'mistral-', 'mistral '
        ],
        'palm': ['palm-', 'palm2', 'palm 2', 'palm-2'],
        'bert': ['bert-base', 'bert base', 'bert-large', 'bert large'],
        'roberta': ['roberta-base', 'roberta base', 'roberta-large', 'roberta large'],
        't5': ['t5-', 't5 ', 't5base', 't5-base', 't5 base']
    }
    
    # Check for other model matches
    for family, variants in base_models.items():
        patterns = [f'^{re.escape(v)}' for v in variants + [family]]
        if any(re.match(pattern, normalized) for pattern in patterns):
            if family in ['llama', 'llama-2', 'mistral']:
                base = family
                
                if any(v in normalized for v in ['instruct', 'ins']):
                    base = f"{base}-instruct"
                elif 'chat' in normalized:
                    base = f"{base}-chat"
                    
                if family == 'mistral' and model_size:
                    return f"{base}-{model_size}", original
                elif model_size:
                    return f"{base}-{model_size}", original
                    
                return base, original
            
            if model_size and family in ['bert', 'roberta', 't5']:
                return f"{family}-{model_size}", original
            
            return family, original
    
    # Remove parenthetical details and extra whitespace
    normalized = re.sub(r'\s*\([^)]*\)\s*', '', normalized)
    normalized = re.sub(r'\s*\[[^\]]*\]\s*', '', normalized)
    normalized = re.sub(r'[-_\s]+', '-', normalized)
    
    if model_size and not normalized.endswith(model_size):
        normalized = f"{normalized}-{model_size}"
        
    return normalized, original

def plot_ridgeline(ax, df, names, counter, colors, vertical_spacing, original_names=None, plot_type=None):
    """Plot the ridgeline distribution"""
    date_min = datetime(2022, 1, 1)
    date_max = datetime(2024, 12, 31)
    
    months = pd.date_range(start=date_min, end=date_max, freq='M')
    month_width = (months[1] - months[0])
    
    monthly_totals = df.groupby(df['date'].dt.to_period('M'))['name'].value_counts()
    max_freq = 0
    
    for idx, name in enumerate(reversed(names)):
        if name in monthly_totals.unstack():
            monthly_counts = monthly_totals.unstack()[name].fillna(0)
            monthly_sums = monthly_totals.groupby('date').sum()
            monthly_freq = monthly_counts / monthly_sums
            max_freq = max(max_freq, monthly_freq.max())
            
            plot_dates = pd.to_datetime([period.to_timestamp() for period in monthly_freq.index])
            baseline = idx * vertical_spacing
            
            # Plot filled bars
            ax.bar(plot_dates, 
                  monthly_freq,
                  width=month_width,
                  bottom=baseline,
                  color=colors[min(idx, len(colors)-1)],
                  alpha=0.6,
                  align='edge',
                  zorder=idx)
            
            # Add border
            ax.bar(plot_dates,
                  monthly_freq,
                  width=month_width,
                  bottom=baseline,
                  color='none',
                  edgecolor=colors[min(idx, len(colors)-1)],
                  linewidth=1.0,
                  alpha=0.9,
                  align='edge',
                  zorder=idx)
            
            # Add peak value label
            max_month_idx = monthly_freq.argmax()
            max_month_date = monthly_freq.index[max_month_idx].to_timestamp()
            ax.text(max_month_date, 
                   baseline + monthly_freq[max_month_idx], 
                   f'{monthly_freq[max_month_idx]:.2f}',
                   ha='left', va='bottom', fontsize=8)
            
            # Add horizontal line
            ax.axhline(y=baseline,
                      color='gray',
                      linewidth=0.2,
                      alpha=0.3,
                      zorder=0)
            
            # Use appropriate display names dictionary based on the plot type
            if plot_type == 'models':
                display_name = model_display_names.get(name, original_names.get(name, name))
            else:  # benchmarks
                display_name = benchmark_display_names.get(name, original_names.get(name, name))
                
            ax.annotate(f"{display_name} ({counter[name]})",
                       xy=(date_max, baseline),
                       xytext=(5, 0),
                       textcoords="offset points",
                       va='center',
                       fontsize=9,
                       fontfamily='serif',
                       fontstyle='normal',
                       bbox=dict(facecolor='white',
                               edgecolor='none',
                               alpha=0.9,
                               pad=1))

def create_combined_plot(df, names, counter, title, filename, color_h, original_names=None, plot_type=None):
    """Create a combined plot with paper counts and distribution"""
    plt.style.use('default')
    plt.rcParams.update({
        'font.family': 'serif',
        'font.size': 10,
        'axes.labelsize': 11,
        'axes.titlesize': 12,
        'figure.titlesize': 14
    })

    fig = plt.figure(figsize=(12, 10))
    
    # Create top subplot for paper counts (25% of height)
    ax1 = plt.subplot2grid((4, 1), (0, 0), rowspan=1)
    
    # Get total mentions per month
    monthly_mentions = df.groupby(df['date'].dt.to_period('M')).size()
    
    start_date = datetime(2022, 1, 1)
    end_date = datetime(2024, 12, 31)
    
    # First, filter papers_df to the date range we want
    papers_df_filtered = papers_df[
        (papers_df.index >= start_date) & 
        (papers_df.index <= end_date)
    ]
    
    # Create date range and align data
    monthly_dates = pd.date_range(start=start_date, end=end_date, freq='M')
    
    # Only include months where we actually have papers
    valid_months = papers_df_filtered.index.to_period('M')
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(valid_months)]
    
    # Resample papers data
    monthly_papers = papers_df_filtered.resample('M')['total_papers'].last()
    
    # Ensure both series align with actual paper months
    monthly_papers = monthly_papers[monthly_papers > 0]
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(monthly_papers.index.to_period('M'))]
    
    # Create date range for plotting
    start_date = datetime(2022, 1, 1)
    end_date = datetime(2024, 12, 31)
    monthly_dates = pd.date_range(start=start_date, end=end_date, freq='MS')  # MS = Month Start
    
    # Filter and align the data
    papers_df_filtered = papers_df[
        (papers_df.index >= start_date) & 
        (papers_df.index <= end_date)
    ]
    
    # Resample to month start frequency
    monthly_papers = papers_df_filtered.resample('MS')['total_papers'].last()
    monthly_mentions = df.groupby(df['date'].dt.to_period('M')).size()
    
    # Ensure both series align with actual paper months
    monthly_papers = monthly_papers[monthly_papers > 0]
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(monthly_papers.index.to_period('M'))]
    
    # Convert period index to timestamp for plotting
    plot_dates = pd.date_range(start=monthly_papers.index.min(), 
                             end=monthly_papers.index.max(), 
                             freq='MS')
    
    # Plot paper counts - align to start of each month
    bars = ax1.bar(plot_dates,
                   monthly_papers.values,
                   width=30,
                   color='gray',
                   alpha=0.5,
                   align='edge')  # Align to start of month

    # Customize top subplot
    ax1.set_xlim(start_date, end_date)
    ax1.spines['top'].set_visible(False)
    ax1.spines['left'].set_visible(False)
    ax1.spines['right'].set_visible(True)
    ax1.yaxis.set_label_position('right')
    ax1.yaxis.tick_right()
    
    # Add value labels for paper counts
    for bar in bars:
        height = bar.get_height()
        if height > 0:
            ax1.text(bar.get_x() + bar.get_width()/2, height,
                    f'{int(height)}',
                    ha='center', va='bottom', fontsize=8)
    
    # Create bottom subplot for distribution (75% of height)
    ax2 = plt.subplot2grid((4, 1), (1, 0), rowspan=3)
    
    n_shades = len(names)
    vertical_spacing = 0.35
    colors = sns.husl_palette(n_shades, h=color_h, s=0.85, l=0.6)
    
    plot_ridgeline(ax2, df, names, counter, colors, vertical_spacing, original_names, plot_type)
    
    ax2.set_xlim(start_date, end_date)
    ax2.yaxis.set_visible(False)
    ax2.spines['top'].set_visible(False)
    ax2.spines['right'].set_visible(False)
    ax2.spines['left'].set_visible(False)
    
    # Format x-axis with year ticks
    for ax in [ax1, ax2]:
        # Set major ticks only at start of year
        ax.xaxis.set_major_locator(mdates.YearLocator(month=1, day=1))
        
        # Remove year labels from ticks
        ax.xaxis.set_major_formatter(plt.NullFormatter())
        
        # Handle year labels differently for each subplot
        if ax == ax1:  # Top subplot
            y_pos = ax.get_ylim()[0] - 0.1
        else:  # Bottom subplot (ax2)
            y_pos = -0.02  # Reduced negative value to move labels up
            
        # Manually add year labels at the middle of each year with larger, bolder font
        years = range(2022, 2025)
        for year in years:
            mid_year_date = datetime(year, 7, 1)
            ax.text(mdates.date2num(mid_year_date), y_pos,
                   str(year),
                   ha='center',
                   va='top',
                   transform=ax.get_xaxis_transform(),
                   fontsize=14,  # Increased font size
                   fontweight='bold')  # Made text bold
        
        # Add minor ticks for all months
        ax.xaxis.set_minor_locator(mdates.MonthLocator())
        
        # Customize tick lengths
        ax.tick_params(axis='x', which='major', length=8)  # Longer ticks for years
        ax.tick_params(axis='x', which='minor', length=4)  # Shorter ticks for months
        
        # Adjust label positions to center them
        ax.tick_params(axis='x', which='major', pad=15)  # Increased padding for year labels
        
        # Add subtle grid
        ax.grid(True, alpha=0.1, linestyle='-', axis='x')
        ax.grid(True, alpha=0.05, linestyle='-', axis='x', which='minor')
    
    # Only show x-label on bottom plot
    ax1.set_xlabel('')
    ax2.set_xlabel('')
    
    # Adjust layout
    plt.subplots_adjust(right=0.85, hspace=0.3)
    
    plt.savefig(filename,
                dpi=300,
                bbox_inches='tight',
                facecolor='white')
    plt.close()

def calculate_statistics(data, start_year=None):
    """Calculate statistics for the dataset with optional year filtering"""
    total_papers = len(set(title for title, paper_info in data.items() 
                          if 'publication_date' in paper_info 
                          and (start_year is None or 
                               datetime.strptime(paper_info['publication_date'], '%Y-%m-%d').year >= start_year)))

    total_model_mentions = 0
    total_benchmark_mentions = 0
    papers_with_models = 0
    papers_with_benchmarks = 0

    for title, paper_info in data.items():
        if 'publication_date' in paper_info and 'analysis' in paper_info:
            date = datetime.strptime(paper_info['publication_date'], '%Y-%m-%d')
            if start_year is None or date.year >= start_year:
                # Count model mentions
                if 'base_models' in paper_info['analysis'] and paper_info['analysis']['base_models']:
                    papers_with_models += 1
                    total_model_mentions += len(paper_info['analysis']['base_models'])
                
                # Count benchmark mentions
                if 'benchmarks' in paper_info['analysis'] and paper_info['analysis']['benchmarks']:
                    papers_with_benchmarks += 1
                    total_benchmark_mentions += len(paper_info['analysis']['benchmarks'])

    # Calculate averages
    avg_models_per_paper = total_model_mentions / total_papers if total_papers > 0 else 0
    avg_benchmarks_per_paper = total_benchmark_mentions / total_papers if total_papers > 0 else 0
    avg_models_when_present = total_model_mentions / papers_with_models if papers_with_models > 0 else 0
    avg_benchmarks_when_present = total_benchmark_mentions / papers_with_benchmarks if papers_with_benchmarks > 0 else 0

    return {
        'total_papers': total_papers,
        'total_model_mentions': total_model_mentions,
        'papers_with_models': papers_with_models,
        'avg_models_per_paper': avg_models_per_paper,
        'avg_models_when_present': avg_models_when_present,
        'total_benchmark_mentions': total_benchmark_mentions,
        'papers_with_benchmarks': papers_with_benchmarks,
        'avg_benchmarks_per_paper': avg_benchmarks_per_paper,
        'avg_benchmarks_when_present': avg_benchmarks_when_present
    }

# Load and process data
with open('papers_data_with_dates.json', 'r') as f:
    papers_data = json.load(f)

# Calculate total papers per month
all_papers_by_month = {}
for title, paper_info in papers_data.items():
    if 'publication_date' in paper_info:
        date = paper_info['publication_date']
        date_obj = datetime.strptime(date, '%Y-%m-%d')
        month_key = date_obj.strftime('%Y-%m')
        all_papers_by_month[month_key] = all_papers_by_month.get(month_key, 0) + 1

# Convert to DataFrame for plotting
papers_df = pd.DataFrame(
    [(datetime.strptime(k, '%Y-%m'), v) for k, v in all_papers_by_month.items()],
    columns=['date', 'total_papers']
)
papers_df.set_index('date', inplace=True)

# Get frequency counts
model_counter = Counter()
benchmark_counter = Counter()
model_original_names = {}
benchmark_original_names = {}
benchmark_records = []
model_records = []

# Process the data
for title, paper_info in papers_data.items():
    if 'publication_date' in paper_info and 'analysis' in paper_info:
        date = paper_info['publication_date']
        date_obj = datetime.strptime(date, '%Y-%m-%d')
        
        # Only process data from 2022 onwards for charts
        if date_obj.year >= 2022:
            # Process benchmarks
            if 'benchmarks' in paper_info['analysis']:
                for benchmark in paper_info['analysis']['benchmarks']:
                    normalized_benchmark = benchmark.strip('"').strip().lower()
                    normalized_benchmark = re.sub(r'[-_\s]+', '-', normalized_benchmark)
                    benchmark_counter[normalized_benchmark] += 1
                    benchmark_original_names[normalized_benchmark] = benchmark.strip('"').strip()
                    benchmark_records.append({
                        'date': date_obj,
                        'name': normalized_benchmark,
                        'type': 'Benchmark',
                        'paper': title
                    })
            
            # Process models
            if 'base_models' in paper_info['analysis']:
                for model in paper_info['analysis']['base_models']:
                    normalized_model, original_name = normalize_model_name(model)
                    model_counter[normalized_model] += 1
                    if normalized_model not in model_original_names:
                        model_original_names[normalized_model] = original_name
                    model_records.append({
                        'date': date_obj,
                        'name': normalized_model,
                        'type': 'Model',
                        'paper': title
                    })

# Get top 20 benchmarks and models
top_20_benchmarks = [b[0] for b in benchmark_counter.most_common(20)]
top_20_models = [m[0] for m in model_counter.most_common(20)]

# Create DataFrames
df_benchmarks = pd.DataFrame(benchmark_records)
df_models = pd.DataFrame(model_records)

# Filter for top 20
df_benchmarks = df_benchmarks[df_benchmarks['name'].isin(top_20_benchmarks)]
df_models = df_models[df_models['name'].isin(top_20_models)]

# Update DataFrames to use datetime
df_benchmarks['date'] = pd.to_datetime(df_benchmarks['date'])
df_models['date'] = pd.to_datetime(df_models['date'])

# Calculate statistics for both full dataset and 2022-2024 period
full_stats = calculate_statistics(papers_data)
recent_stats = calculate_statistics(papers_data, 2022)

# Print statistics for full dataset
print("\nPaper Statistics (Full Dataset):")
print(f"Total papers analyzed: {full_stats['total_papers']}")
print(f"\nModel statistics:")
print(f"Total model mentions: {full_stats['total_model_mentions']}")
print(f"Papers that mention models: {full_stats['papers_with_models']} ({full_stats['papers_with_models']/full_stats['total_papers']*100:.1f}%)")
print(f"Average models per paper (across all papers): {full_stats['avg_models_per_paper']:.2f}")
print(f"Average models when present: {full_stats['avg_models_when_present']:.2f}")

print(f"\nBenchmark statistics:")
print(f"Total benchmark mentions: {full_stats['total_benchmark_mentions']}")
print(f"Papers that mention benchmarks: {full_stats['papers_with_benchmarks']} ({full_stats['papers_with_benchmarks']/full_stats['total_papers']*100:.1f}%)")
print(f"Average benchmarks per paper (across all papers): {full_stats['avg_benchmarks_per_paper']:.2f}")
print(f"Average benchmarks when present: {full_stats['avg_benchmarks_when_present']:.2f}")

# Print statistics for 2022-2024
print("\nPaper Statistics (Jan 2022 - Dec 2024):")
print(f"Total papers analyzed: {recent_stats['total_papers']}")
print(f"\nModel statistics:")
print(f"Total model mentions: {recent_stats['total_model_mentions']}")
print(f"Papers that mention models: {recent_stats['papers_with_models']} ({recent_stats['papers_with_models']/recent_stats['total_papers']*100:.1f}%)")
print(f"Average models per paper (across all papers): {recent_stats['avg_models_per_paper']:.2f}")
print(f"Average models when present: {recent_stats['avg_models_when_present']:.2f}")

print(f"\nBenchmark statistics:")
print(f"Total benchmark mentions: {recent_stats['total_benchmark_mentions']}")
print(f"Papers that mention benchmarks: {recent_stats['papers_with_benchmarks']} ({recent_stats['papers_with_benchmarks']/recent_stats['total_papers']*100:.1f}%)")
print(f"Average benchmarks per paper (across all papers): {recent_stats['avg_benchmarks_per_paper']:.2f}")
print(f"Average benchmarks when present: {recent_stats['avg_benchmarks_when_present']:.2f}")

# Print top 20 lists
print("\nTop 20 Benchmarks and their frequencies:")
for benchmark, count in benchmark_counter.most_common(20):
    original_name = benchmark_original_names.get(benchmark, benchmark)
    print(f"{original_name}: {count}")

print("\nTop 20 Models and their frequencies:")
for model, count in model_counter.most_common(20):
    original_name = model_original_names.get(model, model)
    print(f"{original_name}: {count}")

# Create combined plots
create_combined_plot(df_benchmarks, top_20_benchmarks, benchmark_counter,
                    'Distribution of Benchmark Usage Over Time',
                    'benchmark_distribution.pdf', 0.5,
                    benchmark_original_names)

create_combined_plot(df_models, top_20_models, model_counter,
                    'Distribution of Model Usage Over Time',
                    'model_distribution.pdf', 0.9,
                    model_original_names)
```

```python
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from datetime import datetime
from collections import Counter
import re
import numpy as np
from matplotlib.dates import DateFormatter
import matplotlib.dates as mdates
from scipy import stats

# UPDATED: Standardized model display names with consistent formatting
model_display_names = {
    # GPT models
    'gpt-4': 'GPT-4',
    'gpt-3.5': 'GPT-3.5-Turbo',
    'gpt-3': 'GPT-3',
    'gpt-4o': 'GPT-4o',
    'gpt-4v': 'GPT-4V',
    'gpt-2': 'GPT-2',
    
    # LLaMA models - 1st generation
    'llama': 'LLaMA',
    'llama-7b': 'LLaMA-7B',
    'llama-13b': 'LLaMA-13B',
    'llama-65b': 'LLaMA-65B',
    
    # LLaMA-2 models - consolidated variants
    'llama-2': 'LLaMA-2',
    'llama-2-7b': 'LLaMA-2-7B',  # This will include chat variants
    'llama-2-13b': 'LLaMA-2-13B',  # This will include chat variants
    'llama-2-70b': 'LLaMA-2-70B',  # This will include chat variants
    
    # LLaMA-3 models - consolidated variants
    'llama-3': 'LLaMA-3',
    'llama-3-8b': 'LLaMA-3-8B',  # This will include chat/instruct variants
    'llama-3-70b': 'LLaMA-3-70B',  # This will include chat/instruct variants
    
    # Mistral models - consolidated variants
    'mistral-7b': 'Mistral-7B',  # This will include chat/instruct variants
    'mistral-8x7b': 'Mistral-8x7B',
    
    # Other models
    'palm': 'PaLM-2',
    'palm-2': 'PaLM-2',
    'bert': 'BERT',
    't5': 'T5',
    'roberta': 'RoBERTa',
    
    # Handle any spelling variants (keys must be lowercase)
    'llama3': 'LLaMA-3',
    'llama3-8b': 'LLaMA-3-8B',
    'llama3-70b': 'LLaMA-3-70B',

    # Vicuna
    'vicuna': 'Vicuna',
    'vicuna-13b': 'Vicuna-13B',
    'vicuna-7b': 'Vicuna-7B',
    'vicuna-33b': 'Vicuna-33B',
}

# Dictionary of benchmark display names for nicer chart labels
benchmark_display_names = {
    'mmlu': 'MMLU',
    'truthfulqa': 'TruthfulQA',
    'hellaswag': 'HellaSwag',
    'arc': 'ARC',
    'gsm8k': 'GSM8K',
    'boolq': 'BoolQ',
    'lambada': 'LAMBADA',
    'triviaqa': 'TriviaQA',
    'squad': 'SQuAD',
    'winogrande': 'Winogrande',
    'bbh': 'BBH',
    'math': 'MATH',
    'fever': 'FEVER',
    'openbookqa': 'OpenBookQA',
    'drop': 'DROP',
    'bigbench': 'BIG-bench',
    'glue': 'GLUE',
    'superglue': 'SuperGLUE',
    'natural-questions': 'Natural Questions',
    'c-eval': 'C-Eval',
    'math-qa': 'MathQA',
    'hotpotqa': 'HotpotQA',
    'aqua': 'AQuA',
    'piqa': 'PIQA',
    'hendrycks-test': 'MMLU',
    'nq': 'Natural Questions',
    'commonsenseqa': 'CommonsenseQA',
    'arc-challenge': 'ARC-Challenge',
    'arc-easy': 'ARC-Easy'
}

# Dictionary of model release dates with comprehensive coverage of all variants
model_release_dates = {
    'gpt-4': '2023-03-14',
    'gpt-4v': '2023-09-25',
    'gpt-4o': '2024-05-13',
    'gpt-3.5': '2022-11-30',
    'gpt-3': '2020-06-11',
    'llama': '2023-02-24',
    'llama-7b': '2023-02-24',
    'llama-13b': '2023-02-24',
    'llama-65b': '2023-02-24',
    'llama-2': '2023-07-18',
    'llama-2-7b': '2023-07-18',
    'llama-2-13b': '2023-07-18',
    'llama-2-70b': '2023-07-18',
    'llama-3': '2024-04-18',
    'llama-3-8b': '2024-04-18',
    'llama-3-70b': '2024-04-18',
    'llama-8b': '2024-04-18',  # Additional variant that might be used
    'llama3-8b': '2024-04-18', # Additional variant that might be used
    'llama3': '2024-04-18',    # Additional variant that might be used
    'mistral-7b': '2023-09-25',
    'palm': '2022-04-04',
    'palm-2': '2023-05-10',
    'bert': '2018-10-11',
    'roberta': '2019-07-26',
    't5': '2019-10-23',
    'gpt-2': '2019-02-14',
    'vicuna-13b': '2023-03-13',  # Add Vicuna-13B release date
}

# NEW FUNCTION: Consistently normalize model families by removing chat/instruct variants
def normalize_model_family(name):
    """
    Consistently groups model variants by removing chat/instruct suffixes
    and keeping only the base model family and size
    """
    # Strip any chat/instruct variants
    name = re.sub(r'-(chat|instruct|ins)$', '', name)
    
    # Handle special cases for Llama models, ensure we preserve generation number and size only
    if 'llama' in name:
        # Match llama-[generation]-[size] pattern
        match = re.match(r'(llama-[123])-(\d+[bm])', name)
        if match:
            return f"{match.group(1)}-{match.group(2)}"
        # Handle standalone llama-7b, etc.
        match = re.match(r'(llama)-(\d+[bm])', name)
        if match:
            return f"{match.group(1)}-{match.group(2)}"
    
    # For mistral models
    if 'mistral' in name:
        match = re.match(r'(mistral)-(\d+[bm])', name)
        if match:
            return f"{match.group(1)}-{match.group(2)}"
            
    return name

def normalize_model_name(name: str) -> tuple:
    """
    Normalizes model names while preserving important distinctions between model sizes
    and architectural variants. Returns tuple of (normalized_name, original_name)
    """
    original = name.strip('"').strip()
    normalized = original.lower()
    normalized = re.sub(r'\s+', ' ', normalized)
    
    # Remove quantization information first to avoid it interfering with size detection
    normalized = re.sub(r'\s*\(.*?(?:quantization|quantized|quant).*?\)\s*', '', normalized, flags=re.IGNORECASE)
    
    # Special handling for explicit LLaMA-3 models with complex patterns
    llama3_pattern = r'llama\s*3\s+(\d+[bm])[\s-]*([a-z]*)'
    llama3_match = re.search(llama3_pattern, normalized, re.IGNORECASE)
    if llama3_match:
        size = llama3_match.group(1).lower()
        variant = llama3_match.group(2).lower() if llama3_match.group(2) else ""
        
        if variant:
            return f"llama-3-{size}-{variant}", original
        else:
            return f"llama-3-{size}", original
    
    # Handle special case: "Llama3" (without hyphen) -> "llama-3"
    normalized = re.sub(r'llama(\d+)', r'llama-\1', normalized)
    
    # Extract parameter count/size information - enhanced patterns to catch more variants
    # Try to find size specification like "70B", "7b", "13B", etc.
    size_pattern = r'[\(\[\s]?(\d+[bm])[\)\]\s]?'
    size_match = re.search(size_pattern, normalized, re.IGNORECASE)
    
    # If that fails, look for patterns like "(7 billion parameters)" or "(70 billion)"
    if not size_match:
        billion_pattern = r'[^\d](\d+)[ -]*(billion|b)[^\d]'
        billion_match = re.search(billion_pattern, normalized + ' ', re.IGNORECASE)
        
        if billion_match:
            size = f"{billion_match.group(1)}b"
            model_size = size.lower()
        else:
            # Look for patterns like "(7 million parameters)" or "(70M)"
            million_pattern = r'[^\d](\d+)[ -]*(million|m)[^\d]'
            million_match = re.search(million_pattern, normalized + ' ', re.IGNORECASE)
            
            if million_match:
                size = f"{million_match.group(1)}m"
                model_size = size.lower()
            else:
                model_size = None
    else:
        model_size = size_match.group(1).lower()
    
    # Handle special case for "llama 2/3" which should be split into separate models
    if 'llama 2/3' in normalized or 'llama-2/3' in normalized:
        # We'll default to LLaMA-2 for simplicity
        normalized = 'llama-2'
    
    # Check for chat/instruct variants
    is_chat = any(v in normalized for v in ['chat', '-chat', ' chat'])
    is_instruct = any(v in normalized for v in ['instruct', '-instruct', ' instruct', 'ins'])
    
    # First check for GPT-4 variants
    gpt4_patterns = [
        'gpt4', 'gpt 4', 'gpt-4', 'gpt-4-turbo', 'gpt4-turbo',
        'chatgpt-4', 'chat-gpt-4', 'chatgpt-4.0', 'chat gpt 4', 
        'chatgpt 4', 'gpt-4 turbo', 'gpt 4 turbo'
    ]
    
    if any(re.match(f'^{re.escape(pattern)}', normalized) for pattern in gpt4_patterns):
        if any(v in normalized for v in ['vision', '-v', 'v']):
            return 'gpt-4v', original
        elif any(o in normalized for o in ['4o', '4-o', '4 o']):
            return 'gpt-4o', original
        return 'gpt-4', 'GPT-4'
    
    # Model family mappings with their variants
    base_models = {
        'gpt-3.5': [
            'gpt-3.5-turbo', 'gpt3.5-turbo', 'gpt3.5 turbo', 
            'gpt-3.5-t', 'chatgpt', 'chat-gpt', 'chat gpt', 
            'chatgpt-3.5', 'chatgpt-3.5-turbo', 'chat-gpt-3.5',
            'chat gpt 3.5', 'chatgpt 3.5', 'gpt3.5'
        ],
        'gpt-3': [
            'gpt-3', 'gpt3', 'gpt 3', 'gpt-3-davinci',
            'davinci', 'text-davinci', 'davinci-002',
            'chatgpt-3', 'chat-gpt-3', 'gpt-3-(175b)',
            'gpt-3 (175b)', 'gpt-3 (davinci)'
        ],
        'llama-2': [
            'llama2', 'llama 2', 'llama-2', 'llama2-', 
            'llama-2-', 'llama 2 ', 'llama2 ', 'llama-2 '
        ],
        'llama': [
            'llama1', 'llama 1', 'llama-1', 'llama-', 
            'llama ', 'llamab', 'llama-b'
        ],
        'llama-3': [
            'llama3', 'llama 3', 'llama-3', 'llama3-', 
            'llama-3-', 'llama 3 ', 'llama3 '
        ],
        'mistral': [
            'mistral', 'mistral-', 'mistral '
        ],
        'palm': ['palm-', 'palm2', 'palm 2', 'palm-2'],
        'bert': ['bert-base', 'bert base', 'bert-large', 'bert large'],
        'roberta': ['roberta-base', 'roberta base', 'roberta-large', 'roberta large'],
        't5': ['t5-', 't5 ', 't5base', 't5-base', 't5 base']
    }
    
    # Check for other model matches
    for family, variants in base_models.items():
        patterns = [f'^{re.escape(v)}' for v in variants + [family]]
        if any(re.match(pattern, normalized) for pattern in patterns):
            if family in ['llama', 'llama-2', 'llama-3', 'mistral']:
                base = family
                
                # CONSISTENT HANDLING: Always attach model size first, then variant
                if model_size:
                    # Special handling for LLaMA models without explicit generation number
                    if family == 'llama' and model_size and not any(gen in normalized for gen in ['llama-1', 'llama-2', 'llama-3']):
                        # This is a standalone "llama" with a size but no generation
                        size_without_unit = model_size.rstrip('bm')
                        if size_without_unit.isdigit():
                            # For certain sizes, we can infer the generation
                            if model_size == '8b':
                                # Only LLaMA-3 has 8B models
                                base = 'llama-3'
                            elif model_size == '70b':
                                # For 70B, prefer LLaMA-2 unless context suggests LLaMA-3
                                base = 'llama-2'
                    
                    base = f"{base}-{model_size}"
                    
                # Now add chat/instruct variant consistently
                if is_chat:
                    base = f"{base}-chat"
                elif is_instruct:
                    base = f"{base}-instruct"
                    
                return base, original
            
            if model_size and family in ['bert', 'roberta', 't5']:
                return f"{family}-{model_size}", original
            
            return family, original
    
    # Remove parenthetical details and extra whitespace
    normalized = re.sub(r'\s*\([^)]*\)\s*', '', normalized)
    normalized = re.sub(r'\s*\[[^\]]*\]\s*', '', normalized)
    normalized = re.sub(r'[-_\s]+', '-', normalized)
    
    if model_size and not normalized.endswith(model_size):
        normalized = f"{normalized}-{model_size}"
        
    # Post-processing for LLaMA models with missing generation numbers
    if normalized.startswith('llama-') and not any(gen in normalized for gen in ['llama-1-', 'llama-2-', 'llama-3-']):
        # Extract size if present
        size_match = re.search(r'llama-(\d+[bm])', normalized)
        if size_match:
            size = size_match.group(1)
            # Special case for known LLaMA sizes
            if size == '8b':
                # Only LLaMA-3 has 8B models
                normalized = normalized.replace('llama-8b', 'llama-3-8b')
            elif size == '70b':
                # For 70B, assign to LLaMA-2 by default
                normalized = normalized.replace('llama-70b', 'llama-2-70b')
    
    return normalized, original

def is_anachronistic_extra_check(model_name, paper_date):
    """
    Enhanced anachronism check to catch edge cases.
    Returns True if anachronistic, False otherwise
    """
    # Special handling for LLaMA-3 models - any mention of LLaMA-3 before April 2024 is anachronistic
    if ('llama-3' in model_name or 'llama3' in model_name or 
        'llama-8b' in model_name or model_name == 'llama-8b'):
        llama3_release = datetime.strptime('2024-04-18', '%Y-%m-%d')
        return paper_date < llama3_release
        
    # Check against the model_release_dates dictionary    
    if model_name in model_release_dates:
        model_release = datetime.strptime(model_release_dates[model_name], '%Y-%m-%d')
        return paper_date < model_release
        
    return False  # If we don't know the release date, be conservative

def post_process_model_records(model_records):
    """Apply more aggressive post-processing to fix LLaMA-3 anachronisms and consolidate model variants"""
    filtered_records = []
    
    for record in model_records:
        name = record['name']
        date = record['date']
        
        # Extra check for LLaMA-3 anachronisms
        if is_anachronistic_extra_check(name, date):
            continue  # Skip this record
            
        # Always normalize 8B models to LLaMA-3 WHILE PRESERVING VARIANTS
        if 'llama-8b' in name or ('8b' in name and 'llama' in name and 'llama-3' not in name and 'llama3' not in name):
            # Check for chat/instruct variants before normalizing
            if '-chat' in name:
                name = 'llama-3-8b'  # Now we strip chat variant for consistency
            elif '-instruct' in name:
                name = 'llama-3-8b'  # Now we strip instruct variant for consistency
            else:
                name = 'llama-3-8b'
            record['name'] = name
                
        # NEW: Consistently normalize all model names by stripping chat/instruct variants
        name = normalize_model_family(name)
        record['name'] = name
                
        # Add the record if it passed all checks
        filtered_records.append(record)
    
    return filtered_records

def plot_ridgeline(ax, df, names, counter, colors, vertical_spacing, original_names=None, plot_type=None):
    """Plot the ridgeline distribution"""
    date_min = datetime(2022, 1, 1)
    date_max = datetime(2024, 12, 31)
    
    months = pd.date_range(start=date_min, end=date_max, freq='M')
    month_width = (months[1] - months[0])
    
    monthly_totals = df.groupby(df['date'].dt.to_period('M'))['name'].value_counts()
    max_freq = 0
    
    # Store the baseline y-coordinate for each model so we can add release markers later
    model_baselines = {}
    
    for idx, name in enumerate(reversed(names)):
        if name in monthly_totals.unstack():
            monthly_counts = monthly_totals.unstack()[name].fillna(0)
            monthly_sums = monthly_totals.groupby('date').sum()
            monthly_freq = monthly_counts / monthly_sums
            max_freq = max(max_freq, monthly_freq.max())
            
            plot_dates = pd.to_datetime([period.to_timestamp() for period in monthly_freq.index])
            baseline = idx * vertical_spacing
            
            # Store the baseline for this model for later use
            model_baselines[name] = baseline
            
            # Plot filled bars
            ax.bar(plot_dates, 
                  monthly_freq,
                  width=month_width,
                  bottom=baseline,
                  color=colors[min(idx, len(colors)-1)],
                  alpha=0.6,
                  align='edge',
                  zorder=idx)
            
            # Add border
            ax.bar(plot_dates,
                  monthly_freq,
                  width=month_width,
                  bottom=baseline,
                  color='none',
                  edgecolor=colors[min(idx, len(colors)-1)],
                  linewidth=1.0,
                  alpha=0.9,
                  align='edge',
                  zorder=idx)
            
            # Add peak value label
            max_month_idx = monthly_freq.argmax()
            max_month_date = monthly_freq.index[max_month_idx].to_timestamp()
            ax.text(max_month_date, 
                   baseline + monthly_freq[max_month_idx], 
                   f'{monthly_freq[max_month_idx]:.2f}',
                   ha='left', va='bottom', fontsize=8)
            
            # Add horizontal line
            ax.axhline(y=baseline,
                      color='gray',
                      linewidth=0.2,
                      alpha=0.3,
                      zorder=0)
            
            # Use appropriate display names dictionary based on the plot type
            if plot_type == 'models':
                # Try to find the display name directly or construct one if needed
                if name in model_display_names:
                    display_name = model_display_names[name]
                else:
                    # If not in the dictionary, try to construct a proper display name
                    # based on the normalized name's components
                    parts = name.split('-')
                    if len(parts) >= 2 and parts[0] == 'llama':
                        # Handle LLaMA models
                        if len(parts) == 2:  # llama-2, llama-3
                            display_name = f"LLaMA-{parts[1].upper()}"
                        elif len(parts) == 3:  # llama-2-7b, llama-3-70b
                            display_name = f"LLaMA-{parts[1]}-{parts[2].upper()}"
                        elif len(parts) >= 4:  # llama-2-7b-chat, llama-3-8b-instruct
                            suffix = parts[3].capitalize() if len(parts) > 3 else ""
                            display_name = f"LLaMA-{parts[1]}-{parts[2].upper()}-{suffix}"
                    elif len(parts) >= 2 and parts[0] == 'mistral':
                        # Handle Mistral models
                        if len(parts) == 2:  # mistral-7b
                            display_name = f"Mistral-{parts[1].upper()}"
                        elif len(parts) >= 3:  # mistral-7b-instruct
                            suffix = parts[2].capitalize() if len(parts) > 2 else ""
                            display_name = f"Mistral-{parts[1].upper()}-{suffix}"
                    else:
                        # Use original name or the name as is
                        display_name = original_names.get(name, name.upper())
            else:  # benchmarks
                display_name = benchmark_display_names.get(name, original_names.get(name, name))
                
            ax.annotate(f"{display_name} ({counter[name]})",
                       xy=(date_max, baseline),
                       xytext=(5, 0),
                       textcoords="offset points",
                       va='center',
                       fontsize=9,
                       fontfamily='serif',
                       fontstyle='normal',
                       bbox=dict(facecolor='white',
                               edgecolor='none',
                               alpha=0.9,
                               pad=1))
    
    # Return the model baselines for adding release markers later
    return model_baselines

def create_combined_plot(df, names, counter, title, filename, color_h, original_names=None, plot_type=None):
    """Create a combined plot with paper counts and distribution"""
    plt.style.use('default')
    plt.rcParams.update({
        'font.family': 'serif',
        'font.size': 10,
        'axes.labelsize': 11,
        'axes.titlesize': 12,
        'figure.titlesize': 14
    })

    fig = plt.figure(figsize=(12, 10))
    
    # Create top subplot for paper counts (25% of height)
    ax1 = plt.subplot2grid((4, 1), (0, 0), rowspan=1)
    
    # Get total mentions per month
    monthly_mentions = df.groupby(df['date'].dt.to_period('M')).size()
    
    start_date = datetime(2022, 1, 1)
    end_date = datetime(2024, 12, 31)
    
    # First, filter papers_df to the date range we want
    papers_df_filtered = papers_df[
        (papers_df.index >= start_date) & 
        (papers_df.index <= end_date)
    ]
    
    # Create date range and align data
    monthly_dates = pd.date_range(start=start_date, end=end_date, freq='M')
    
    # Only include months where we actually have papers
    valid_months = papers_df_filtered.index.to_period('M')
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(valid_months)]
    
    # Resample papers data
    monthly_papers = papers_df_filtered.resample('M')['total_papers'].last()
    
    # Ensure both series align with actual paper months
    monthly_papers = monthly_papers[monthly_papers > 0]
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(monthly_papers.index.to_period('M'))]
    
    # Create date range for plotting
    start_date = datetime(2022, 1, 1)
    end_date = datetime(2024, 12, 31)
    monthly_dates = pd.date_range(start=start_date, end=end_date, freq='MS')  # MS = Month Start
    
    # Filter and align the data
    papers_df_filtered = papers_df[
        (papers_df.index >= start_date) & 
        (papers_df.index <= end_date)
    ]
    
    # Resample to month start frequency
    monthly_papers = papers_df_filtered.resample('MS')['total_papers'].last()
    monthly_mentions = df.groupby(df['date'].dt.to_period('M')).size()
    
    # Ensure both series align with actual paper months
    monthly_papers = monthly_papers[monthly_papers > 0]
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(monthly_papers.index.to_period('M'))]
    
    # Convert period index to timestamp for plotting
    plot_dates = pd.date_range(start=monthly_papers.index.min(), 
                             end=monthly_papers.index.max(), 
                             freq='MS')
    
    # Plot paper counts - align to start of each month
    bars = ax1.bar(plot_dates,
                   monthly_papers.values,
                   width=30,
                   color='gray',
                   alpha=0.5,
                   align='edge')  # Align to start of month

    # Customize top subplot
    ax1.set_xlim(start_date, end_date)
    ax1.spines['top'].set_visible(False)
    ax1.spines['left'].set_visible(False)
    ax1.spines['right'].set_visible(True)
    ax1.yaxis.set_label_position('right')
    ax1.yaxis.tick_right()
    
    # Add value labels for paper counts
    for bar in bars:
        height = bar.get_height()
        if height > 0:
            ax1.text(bar.get_x() + bar.get_width()/2, height,
                    f'{int(height)}',
                    ha='center', va='bottom', fontsize=8)
    
    # Create bottom subplot for distribution (75% of height)
    ax2 = plt.subplot2grid((4, 1), (1, 0), rowspan=3)
    
    n_shades = len(names)
    vertical_spacing = 0.35
    colors = sns.husl_palette(n_shades, h=color_h, s=0.85, l=0.6)
    
    # Plot ridgeline and get the model baselines
    model_baselines = plot_ridgeline(ax2, df, names, counter, colors, vertical_spacing, original_names, plot_type)
    
    # Add release date markers for each model if this is a models plot
    if plot_type == 'models':
        for name in names:
            # Check if this model has a release date
            if name in model_release_dates:
                release_date = datetime.strptime(model_release_dates[name], '%Y-%m-%d')
                
                # Check if the release date is within our plot range
                if start_date <= release_date <= end_date:
                    # Get the baseline y-coordinate for this model
                    baseline = model_baselines.get(name, 0)
                    
                    # Add a marker at the release date - just a simple tick mark
                    ax2.scatter([release_date], [baseline], 
                               marker='|',  # Vertical line marker
                               s=100,  # Size
                               color='black',  # Black color for visibility
                               linewidth=1.5,  # Thicker line
                               zorder=100)  # Ensure it's on top
    
    ax2.set_xlim(start_date, end_date)
    ax2.yaxis.set_visible(False)
    ax2.spines['top'].set_visible(False)
    ax2.spines['right'].set_visible(False)
    ax2.spines['left'].set_visible(False)
    
    # Format x-axis with year ticks
    for ax in [ax1, ax2]:
        # Set major ticks only at start of year
        ax.xaxis.set_major_locator(mdates.YearLocator(month=1, day=1))
        
        # Remove year labels from ticks
        ax.xaxis.set_major_formatter(plt.NullFormatter())
        
        # Handle year labels differently for each subplot
        if ax == ax1:  # Top subplot
            y_pos = ax.get_ylim()[0] - 0.1
        else:  # Bottom subplot (ax2)
            y_pos = -0.02  # Reduced negative value to move labels up
            
        # Manually add year labels at the middle of each year with larger, bolder font
        years = range(2022, 2025)
        for year in years:
            mid_year_date = datetime(year, 7, 1)
            ax.text(mdates.date2num(mid_year_date), y_pos,
                   str(year),
                   ha='center',
                   va='top',
                   transform=ax.get_xaxis_transform(),
                   fontsize=14,  # Increased font size
                   fontweight='bold')  # Made text bold
        
        # Add minor ticks for all months
        ax.xaxis.set_minor_locator(mdates.MonthLocator())
        
        # Customize tick lengths
        ax.tick_params(axis='x', which='major', length=8)  # Longer ticks for years
        ax.tick_params(axis='x', which='minor', length=4)  # Shorter ticks for months
        
        # Adjust label positions to center them
        ax.tick_params(axis='x', which='major', pad=15)  # Increased padding for year labels
        
        # Add subtle grid
        ax.grid(True, alpha=0.1, linestyle='-', axis='x')
        ax.grid(True, alpha=0.05, linestyle='-', axis='x', which='minor')
    
    # Only show x-label on bottom plot
    ax1.set_xlabel('')
    ax2.set_xlabel('')
    
    # Adjust layout
    plt.subplots_adjust(right=0.85, hspace=0.3)
    
    plt.savefig(filename,
                dpi=300,
                bbox_inches='tight',
                facecolor='white')
    plt.close()

def analyze_and_report_anachronisms(papers_data):
    """Run the analysis of anachronistic model mentions and print a report"""
    problematic_papers = []
    
    for title, paper_info in papers_data.items():
        if 'publication_date' in paper_info and 'analysis' in paper_info:
            paper_date = datetime.strptime(paper_info['publication_date'], '%Y-%m-%d')
            
            anachronistic_models = []
            if 'base_models' in paper_info['analysis']:
                for model in paper_info['analysis']['base_models']:
                    normalized_model, _ = normalize_model_name(model)
                    
                    # Use enhanced anachronism check
                    if is_anachronistic_extra_check(normalized_model, paper_date):
                        anachronistic_models.append({
                            'model': normalized_model,
                            'paper_date': paper_date.strftime('%Y-%m-%d'),
                            'model_release_date': model_release_dates.get(normalized_model, 'Unknown'),
                            'days_difference': (datetime.strptime(model_release_dates.get(normalized_model, '2099-01-01'), '%Y-%m-%d') - paper_date).days if normalized_model in model_release_dates else 0
                        })
            
            if anachronistic_models:
                problematic_papers.append({
                    'title': title,
                    'publication_date': paper_info['publication_date'],
                    'anachronistic_models': anachronistic_models
                })
    
    print(f"\n{'='*80}")
    print(f"ANACHRONISTIC MODEL MENTIONS REPORT")
    print(f"{'='*80}")
    
    if not problematic_papers:
        print("No anachronistic model mentions found.")
    else:
        print(f"Found {len(problematic_papers)} papers with anachronistic model mentions.")
        print(f"These papers mention models that weren't released at the time of publication:\n")
        
        for i, paper in enumerate(problematic_papers, 1):
            print(f"{i}. Paper: {paper['title']}")
            print(f"   Published: {paper['publication_date']}")
            print(f"   Anachronistic models mentioned:")
            
            for model_info in paper['anachronistic_models']:
                release_date = model_info['model_release_date']
                if release_date != 'Unknown':
                    print(f"   - {model_info['model']} (released {release_date}, "
                          f"{model_info['days_difference']} days after paper publication)")
                else:
                    print(f"   - {model_info['model']} (release date unknown)")
            
            print()
    
    return problematic_papers

# Main execution
if __name__ == "__main__":
    # Test problematic model names for debugging
    test_models = [
        "Llama-3-8B",
        "Llama3 (70B)",
        "Llama 3 8B-instruct (4-bit quantization)",
        "Mistral 7B",
        "LLaMA-2 (70 billion parameters)",
        "LLaMA 70B",  # This should be LLaMA-2-70B
        "LLaMA 8B",   # This should be LLaMA-3-8B
    ]
    
    print("\nTesting model name normalization and display:")
    for model in test_models:
        normalized, original = normalize_model_name(model)
        normalized = normalize_model_family(normalized)  # NEW: Apply family normalization to strip chat/instruct
        display = model_display_names.get(normalized, normalized)
        print(f"Original: '{model}' -> Normalized: '{normalized}' -> Display: '{display}'")
    
    # Load and process data
    with open('papers_data_with_dates.json', 'r') as f:
        papers_data = json.load(f)

    # Calculate total papers per month
    all_papers_by_month = {}
    for title, paper_info in papers_data.items():
        if 'publication_date' in paper_info:
            date = paper_info['publication_date']
            date_obj = datetime.strptime(date, '%Y-%m-%d')
            month_key = date_obj.strftime('%Y-%m')
            all_papers_by_month[month_key] = all_papers_by_month.get(month_key, 0) + 1

    # Convert to DataFrame for plotting
    papers_df = pd.DataFrame(
        [(datetime.strptime(k, '%Y-%m'), v) for k, v in all_papers_by_month.items()],
        columns=['date', 'total_papers']
    )
    papers_df.set_index('date', inplace=True)

    # Get frequency counts
    model_counter = Counter()
    benchmark_counter = Counter()
    model_original_names = {}
    benchmark_original_names = {}
    benchmark_records = []
    model_records = []

    # Look for models that don't have a size (like standalone "LLaMA-2")
    standalone_models = set()

    # Process the data - FILTER OUT ANACHRONISTIC MENTIONS
    for title, paper_info in papers_data.items():
        if 'publication_date' in paper_info and 'analysis' in paper_info:
            date = paper_info['publication_date']
            date_obj = datetime.strptime(date, '%Y-%m-%d')
            
            # Only process data from 2022 onwards for charts
            if date_obj.year >= 2022:
                # Process benchmarks (no anachronism filtering for benchmarks)
                if 'benchmarks' in paper_info['analysis']:
                    for benchmark in paper_info['analysis']['benchmarks']:
                        normalized_benchmark = benchmark.strip('"').strip().lower()
                        normalized_benchmark = re.sub(r'[-_\s]+', '-', normalized_benchmark)
                        benchmark_counter[normalized_benchmark] += 1
                        benchmark_original_names[normalized_benchmark] = benchmark.strip('"').strip()
                        benchmark_records.append({
                            'date': date_obj,
                            'name': normalized_benchmark,
                            'type': 'Benchmark',
                            'paper': title
                        })
                
                # Process models - SKIP ANACHRONISTIC MENTIONS with enhanced checks
                if 'base_models' in paper_info['analysis']:
                    for model in paper_info['analysis']['base_models']:
                        normalized_model, original_name = normalize_model_name(model)
                        
                        # Enhanced anachronism check - more aggressive for LLaMA-3
                        if is_anachronistic_extra_check(normalized_model, date_obj):
                            continue
                        
                        # NEW: Apply consistent family normalization to strip chat/instruct variants
                        normalized_model = normalize_model_family(normalized_model)
                        
                        # Check for models without size specifications
                        if normalized_model in ['llama-2', 'llama-3'] and not re.search(r'\d+[bm]', normalized_model):
                            standalone_models.add((normalized_model, original_name))
                        
                        model_counter[normalized_model] += 1
                        if normalized_model not in model_original_names:
                            model_original_names[normalized_model] = original_name
                        model_records.append({
                            'date': date_obj,
                            'name': normalized_model,
                            'type': 'Model',
                            'paper': title
                        })

    # Apply more aggressive post-processing to fix any remaining LLaMA-3 issues
    model_records = post_process_model_records(model_records)
    
    # Rebuild the counter after post-processing
    model_counter = Counter([record['name'] for record in model_records])

    # Print any standalone models found (for investigation)
    if standalone_models:
        print("\nStandalone models without size specifications found:")
        for model, original in standalone_models:
            print(f"  - Normalized: '{model}', Original: '{original}'")

    # Get top 20 benchmarks and models
    top_20_benchmarks = [b[0] for b in benchmark_counter.most_common(20)]
    top_20_models = [m[0] for m in model_counter.most_common(20)]

    # Create DataFrames
    df_benchmarks = pd.DataFrame(benchmark_records)
    df_models = pd.DataFrame(model_records)

    # Filter for top 20
    df_benchmarks = df_benchmarks[df_benchmarks['name'].isin(top_20_benchmarks)]
    df_models = df_models[df_models['name'].isin(top_20_models)]

    # Update DataFrames to use datetime
    df_benchmarks['date'] = pd.to_datetime(df_benchmarks['date'])
    df_models['date'] = pd.to_datetime(df_models['date'])

    # Run anachronism analysis and report
    problematic_papers = analyze_and_report_anachronisms(papers_data)
    
    # Save the anachronistic mentions to a file
    with open('anachronistic_mentions.json', 'w') as f:
        json.dump(problematic_papers, f, indent=2)

    # Create combined plots
    create_combined_plot(df_benchmarks, top_20_benchmarks, benchmark_counter,
                        'Distribution of Benchmark Usage Over Time',
                        'benchmark_distribution_v2.pdf', 0.5,
                        benchmark_original_names,
                        'benchmarks')

    create_combined_plot(df_models, top_20_models, model_counter,
                        'Distribution of Model Usage Over Time',
                        'model_distribution_v2.pdf', 0.9,
                        model_original_names,
                        'models')
```

```python
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from datetime import datetime
from collections import Counter
import re
import numpy as np
from matplotlib.dates import DateFormatter
import matplotlib.dates as mdates
from scipy import stats

# UPDATED: Standardized model display names with consistent formatting
model_display_names = {
    # GPT models
    'gpt-4': 'GPT-4',
    'gpt-3.5': 'GPT-3.5-Turbo',
    'gpt-3': 'GPT-3',
    'gpt-4o': 'GPT-4o',
    'gpt-4v': 'GPT-4V',
    'gpt-2': 'GPT-2',
    
    # LLaMA models - 1st generation
    'llama': 'LLaMA',
    'llama-7b': 'LLaMA-7B',
    'llama-13b': 'LLaMA-13B',
    'llama-65b': 'LLaMA-65B',
    'llama-70b': 'LLaMA-70B',
    
    # LLaMA-2 models - consolidated variants
    'llama-2': 'LLaMA-2',
    'llama-2-7b': 'LLaMA-2-7B',  # This will include chat variants
    'llama-2-13b': 'LLaMA-2-13B',  # This will include chat variants
    'llama-2-70b': 'LLaMA-2-70B',  # This will include chat variants
    
    # LLaMA-3 models - consolidated variants
    'llama-3': 'LLaMA-3',
    'llama-3-8b': 'LLaMA-3-8B',  # This will include chat/instruct variants
    'llama-3-70b': 'LLaMA-3-70B',  # This will include chat/instruct variants
    
    # Mistral models - consolidated variants
    'mistral-7b': 'Mistral-7B',  # This will include chat/instruct variants
    'mistral-8x7b': 'Mistral-8x7B',
    
    # Other models
    'palm': 'PaLM-2',
    'palm-2': 'PaLM-2',
    'bert': 'BERT',
    't5': 'T5',
    'roberta': 'RoBERTa',
    'vicuna-13b': 'Vicuna-13B',
    'vicuna-7b': 'Vicuna-7B',
    
    # Handle any spelling variants (keys must be lowercase)
    'llama3': 'LLaMA-3',
    'llama3-8b': 'LLaMA-3-8B',
    'llama3-70b': 'LLaMA-3-70B',
}

# Dictionary of benchmark display names for nicer chart labels
benchmark_display_names = {
    'mmlu': 'MMLU',
    'truthfulqa': 'TruthfulQA',
    'hellaswag': 'HellaSwag',
    'arc': 'ARC',
    'gsm8k': 'GSM8K',
    'boolq': 'BoolQ',
    'lambada': 'LAMBADA',
    'triviaqa': 'TriviaQA',
    'squad': 'SQuAD',
    'winogrande': 'Winogrande',
    'bbh': 'BBH',
    'math': 'MATH',
    'fever': 'FEVER',
    'openbookqa': 'OpenBookQA',
    'drop': 'DROP',
    'bigbench': 'BIG-bench',
    'glue': 'GLUE',
    'superglue': 'SuperGLUE',
    'natural-questions': 'Natural Questions',
    'c-eval': 'C-Eval',
    'math-qa': 'MathQA',
    'hotpotqa': 'HotpotQA',
    'aqua': 'AQuA',
    'piqa': 'PIQA',
    'hendrycks-test': 'MMLU',
    'nq': 'Natural Questions',
    'commonsenseqa': 'CommonsenseQA',
    'arc-challenge': 'ARC-Challenge',
    'arc-easy': 'ARC-Easy',
    'strategyqa': 'StrategyQA',
    'svamp': 'SVAMP',
    'humaneval': 'HumanEval',
    'mbpp': 'MBPP',
    'multiarith': 'MultiArith',
    'asdiv': 'ASDiv'
}

# Dictionary of model release dates with comprehensive coverage of all variants
model_release_dates = {
    'gpt-4': '2023-03-14',
    'gpt-4v': '2023-09-25',
    'gpt-4o': '2024-05-13',
    'gpt-3.5': '2022-11-30',
    'gpt-3': '2020-06-11',
    'llama': '2023-02-24',
    'llama-7b': '2023-02-24',
    'llama-13b': '2023-02-24',
    'llama-65b': '2023-02-24',
    'llama-70b': '2023-02-24',
    'llama-2': '2023-07-18',
    'llama-2-7b': '2023-07-18',
    'llama-2-13b': '2023-07-18',
    'llama-2-70b': '2023-07-18',
    'llama-3': '2024-04-18',
    'llama-3-8b': '2024-04-18',
    'llama-3-70b': '2024-04-18',
    'llama-8b': '2024-04-18',  # Additional variant that might be used
    'llama3-8b': '2024-04-18', # Additional variant that might be used
    'llama3': '2024-04-18',    # Additional variant that might be used
    'mistral-7b': '2023-09-25',
    'palm': '2022-04-04',
    'palm-2': '2023-05-10',
    'bert': '2018-10-11',
    'roberta': '2019-07-26',
    't5': '2019-10-23',
    'gpt-2': '2019-02-14',
    'vicuna-13b': '2023-03-13',  # Added Vicuna-13B release date
}

# NEW FUNCTION: Consistently normalize model families by removing chat/instruct variants
def normalize_model_family(name):
    """
    Consistently groups model variants by removing chat/instruct suffixes
    and keeping only the base model family and size
    """
    # Strip any chat/instruct variants
    name = re.sub(r'-(chat|instruct|ins)$', '', name)
    
    # Handle special cases for Llama models, ensure we preserve generation number and size only
    if 'llama' in name:
        # Match llama-[generation]-[size] pattern
        match = re.match(r'(llama-[123])-(\d+[bm])', name)
        if match:
            return f"{match.group(1)}-{match.group(2)}"
        # Handle standalone llama-7b, etc.
        match = re.match(r'(llama)-(\d+[bm])', name)
        if match:
            return f"{match.group(1)}-{match.group(2)}"
    
    # For mistral models
    if 'mistral' in name:
        match = re.match(r'(mistral)-(\d+[bm])', name)
        if match:
            return f"{match.group(1)}-{match.group(2)}"
            
    return name

def normalize_model_name(name: str) -> tuple:
    """
    Normalizes model names while preserving important distinctions between model sizes
    and architectural variants. Returns tuple of (normalized_name, original_name)
    """
    original = name.strip('"').strip()
    normalized = original.lower()
    normalized = re.sub(r'\s+', ' ', normalized)
    
    # Remove quantization information first to avoid it interfering with size detection
    normalized = re.sub(r'\s*\(.*?(?:quantization|quantized|quant).*?\)\s*', '', normalized, flags=re.IGNORECASE)
    
    # Special handling for explicit LLaMA-3 models with complex patterns
    llama3_pattern = r'llama\s*3\s+(\d+[bm])[\s-]*([a-z]*)'
    llama3_match = re.search(llama3_pattern, normalized, re.IGNORECASE)
    if llama3_match:
        size = llama3_match.group(1).lower()
        variant = llama3_match.group(2).lower() if llama3_match.group(2) else ""
        
        if variant:
            return f"llama-3-{size}-{variant}", original
        else:
            return f"llama-3-{size}", original
    
    # Handle special case: "Llama3" (without hyphen) -> "llama-3"
    normalized = re.sub(r'llama(\d+)', r'llama-\1', normalized)
    
    # Extract parameter count/size information - enhanced patterns to catch more variants
    # Try to find size specification like "70B", "7b", "13B", etc.
    size_pattern = r'[\(\[\s]?(\d+[bm])[\)\]\s]?'
    size_match = re.search(size_pattern, normalized, re.IGNORECASE)
    
    # If that fails, look for patterns like "(7 billion parameters)" or "(70 billion)"
    if not size_match:
        billion_pattern = r'[^\d](\d+)[ -]*(billion|b)[^\d]'
        billion_match = re.search(billion_pattern, normalized + ' ', re.IGNORECASE)
        
        if billion_match:
            size = f"{billion_match.group(1)}b"
            model_size = size.lower()
        else:
            # Look for patterns like "(7 million parameters)" or "(70M)"
            million_pattern = r'[^\d](\d+)[ -]*(million|m)[^\d]'
            million_match = re.search(million_pattern, normalized + ' ', re.IGNORECASE)
            
            if million_match:
                size = f"{million_match.group(1)}m"
                model_size = size.lower()
            else:
                model_size = None
    else:
        model_size = size_match.group(1).lower()
    
    # Handle special case for "llama 2/3" which should be split into separate models
    if 'llama 2/3' in normalized or 'llama-2/3' in normalized:
        # We'll default to LLaMA-2 for simplicity
        normalized = 'llama-2'
    
    # Check for chat/instruct variants
    is_chat = any(v in normalized for v in ['chat', '-chat', ' chat'])
    is_instruct = any(v in normalized for v in ['instruct', '-instruct', ' instruct', 'ins'])
    
    # First check for GPT-4 variants
    gpt4_patterns = [
        'gpt4', 'gpt 4', 'gpt-4', 'gpt-4-turbo', 'gpt4-turbo',
        'chatgpt-4', 'chat-gpt-4', 'chatgpt-4.0', 'chat gpt 4', 
        'chatgpt 4', 'gpt-4 turbo', 'gpt 4 turbo'
    ]
    
    if any(re.match(f'^{re.escape(pattern)}', normalized) for pattern in gpt4_patterns):
        if any(v in normalized for v in ['vision', '-v', 'v']):
            return 'gpt-4v', original
        elif any(o in normalized for o in ['4o', '4-o', '4 o']):
            return 'gpt-4o', original
        return 'gpt-4', 'GPT-4'
    
    # Model family mappings with their variants
    base_models = {
        'gpt-3.5': [
            'gpt-3.5-turbo', 'gpt3.5-turbo', 'gpt3.5 turbo', 
            'gpt-3.5-t', 'chatgpt', 'chat-gpt', 'chat gpt', 
            'chatgpt-3.5', 'chatgpt-3.5-turbo', 'chat-gpt-3.5',
            'chat gpt 3.5', 'chatgpt 3.5', 'gpt3.5'
        ],
        'gpt-3': [
            'gpt-3', 'gpt3', 'gpt 3', 'gpt-3-davinci',
            'davinci', 'text-davinci', 'davinci-002',
            'chatgpt-3', 'chat-gpt-3', 'gpt-3-(175b)',
            'gpt-3 (175b)', 'gpt-3 (davinci)'
        ],
        'llama-2': [
            'llama2', 'llama 2', 'llama-2', 'llama2-', 
            'llama-2-', 'llama 2 ', 'llama2 ', 'llama-2 '
        ],
        'llama': [
            'llama1', 'llama 1', 'llama-1', 'llama-', 
            'llama ', 'llamab', 'llama-b'
        ],
        'llama-3': [
            'llama3', 'llama 3', 'llama-3', 'llama3-', 
            'llama-3-', 'llama 3 ', 'llama3 '
        ],
        'mistral': [
            'mistral', 'mistral-', 'mistral '
        ],
        'vicuna': [
            'vicuna', 'vicuna-', 'vicuna '
        ],
        'palm': ['palm-', 'palm2', 'palm 2', 'palm-2'],
        'bert': ['bert-base', 'bert base', 'bert-large', 'bert large'],
        'roberta': ['roberta-base', 'roberta base', 'roberta-large', 'roberta large'],
        't5': ['t5-', 't5 ', 't5base', 't5-base', 't5 base']
    }
    
    # Check for other model matches
    for family, variants in base_models.items():
        patterns = [f'^{re.escape(v)}' for v in variants + [family]]
        if any(re.match(pattern, normalized) for pattern in patterns):
            if family in ['llama', 'llama-2', 'llama-3', 'mistral', 'vicuna']:
                base = family
                
                # CONSISTENT HANDLING: Always attach model size first, then variant
                if model_size:
                    # Special handling for LLaMA models without explicit generation number
                    if family == 'llama' and model_size and not any(gen in normalized for gen in ['llama-1', 'llama-2', 'llama-3']):
                        # This is a standalone "llama" with a size but no generation
                        size_without_unit = model_size.rstrip('bm')
                        if size_without_unit.isdigit():
                            # For certain sizes, we can infer the generation
                            if model_size == '8b':
                                # Only LLaMA-3 has 8B models
                                base = 'llama-3'
                            elif model_size == '70b':
                                # For 70B, prefer LLaMA-2 unless context suggests LLaMA-3
                                base = 'llama-2'
                    
                    base = f"{base}-{model_size}"
                    
                # Now add chat/instruct variant consistently
                if is_chat:
                    base = f"{base}-chat"
                elif is_instruct:
                    base = f"{base}-instruct"
                    
                return base, original
            
            if model_size and family in ['bert', 'roberta', 't5']:
                return f"{family}-{model_size}", original
            
            return family, original
    
    # Remove parenthetical details and extra whitespace
    normalized = re.sub(r'\s*\([^)]*\)\s*', '', normalized)
    normalized = re.sub(r'\s*\[[^\]]*\]\s*', '', normalized)
    normalized = re.sub(r'[-_\s]+', '-', normalized)
    
    if model_size and not normalized.endswith(model_size):
        normalized = f"{normalized}-{model_size}"
        
    # Post-processing for LLaMA models with missing generation numbers
    if normalized.startswith('llama-') and not any(gen in normalized for gen in ['llama-1-', 'llama-2-', 'llama-3-']):
        # Extract size if present
        size_match = re.search(r'llama-(\d+[bm])', normalized)
        if size_match:
            size = size_match.group(1)
            # Special case for known LLaMA sizes
            if size == '8b':
                # Only LLaMA-3 has 8B models
                normalized = normalized.replace('llama-8b', 'llama-3-8b')
            elif size == '70b':
                # For 70B, assign to LLaMA-2 by default
                normalized = normalized.replace('llama-70b', 'llama-2-70b')
    
    return normalized, original

def is_anachronistic_extra_check(model_name, paper_date):
    """
    Enhanced anachronism check to catch edge cases.
    Returns True if anachronistic, False otherwise
    """
    # Special handling for LLaMA-3 models - any mention of LLaMA-3 before April 2024 is anachronistic
    if ('llama-3' in model_name or 'llama3' in model_name or 
        'llama-8b' in model_name or model_name == 'llama-8b'):
        llama3_release = datetime.strptime('2024-04-18', '%Y-%m-%d')
        return paper_date < llama3_release
        
    # Check against the model_release_dates dictionary    
    if model_name in model_release_dates:
        model_release = datetime.strptime(model_release_dates[model_name], '%Y-%m-%d')
        return paper_date < model_release
        
    return False  # If we don't know the release date, be conservative

def post_process_model_records(model_records):
    """Apply more aggressive post-processing to fix LLaMA-3 anachronisms and consolidate model variants"""
    filtered_records = []
    
    for record in model_records:
        name = record['name']
        date = record['date']
        
        # Extra check for LLaMA-3 anachronisms
        if is_anachronistic_extra_check(name, date):
            continue  # Skip this record
            
        # Always normalize 8B models to LLaMA-3 WHILE PRESERVING VARIANTS
        if 'llama-8b' in name or ('8b' in name and 'llama' in name and 'llama-3' not in name and 'llama3' not in name):
            # Check for chat/instruct variants before normalizing
            if '-chat' in name:
                name = 'llama-3-8b'  # Now we strip chat variant for consistency
            elif '-instruct' in name:
                name = 'llama-3-8b'  # Now we strip instruct variant for consistency
            else:
                name = 'llama-3-8b'
            record['name'] = name
                
        # NEW: Consistently normalize all model names by stripping chat/instruct variants
        name = normalize_model_family(name)
        record['name'] = name
                
        # Add the record if it passed all checks
        filtered_records.append(record)
    
    return filtered_records

def plot_ridgeline(ax, df, names, counter, colors, vertical_spacing, original_names=None, plot_type=None):
    """Plot the ridgeline distribution"""
    date_min = datetime(2022, 1, 1)
    date_max = datetime(2024, 12, 31)
    
    months = pd.date_range(start=date_min, end=date_max, freq='M')
    month_width = (months[1] - months[0])
    
    monthly_totals = df.groupby(df['date'].dt.to_period('M'))['name'].value_counts()
    max_freq = 0
    
    # Store the baseline y-coordinate for each model so we can add release markers later
    model_baselines = {}
    
    for idx, name in enumerate(reversed(names)):
        if name in monthly_totals.unstack():
            monthly_counts = monthly_totals.unstack()[name].fillna(0)
            monthly_sums = monthly_totals.groupby('date').sum()
            monthly_freq = monthly_counts / monthly_sums
            max_freq = max(max_freq, monthly_freq.max())
            
            plot_dates = pd.to_datetime([period.to_timestamp() for period in monthly_freq.index])
            baseline = idx * vertical_spacing
            
            # Store the baseline for this model for later use
            model_baselines[name] = baseline
            
            # Plot filled bars
            ax.bar(plot_dates, 
                  monthly_freq,
                  width=month_width,
                  bottom=baseline,
                  color=colors[min(idx, len(colors)-1)],
                  alpha=0.6,
                  align='edge',
                  zorder=idx)
            
            # Add border
            ax.bar(plot_dates,
                  monthly_freq,
                  width=month_width,
                  bottom=baseline,
                  color='none',
                  edgecolor=colors[min(idx, len(colors)-1)],
                  linewidth=1.0,
                  alpha=0.9,
                  align='edge',
                  zorder=idx)
            
            # Add peak value label
            max_month_idx = monthly_freq.argmax()
            max_month_date = monthly_freq.index[max_month_idx].to_timestamp()
            ax.text(max_month_date, 
                   baseline + monthly_freq[max_month_idx], 
                   f'{monthly_freq[max_month_idx]:.2f}',
                   ha='left', va='bottom', fontsize=8)
            
            # Add horizontal line
            ax.axhline(y=baseline,
                      color='gray',
                      linewidth=0.2,
                      alpha=0.3,
                      zorder=0)
            
            # Use appropriate display names dictionary based on the plot type
            if plot_type == 'models':
                # Try to find the display name directly or construct one if needed
                if name in model_display_names:
                    display_name = model_display_names[name]
                else:
                    # If not in the dictionary, try to construct a proper display name
                    # based on the normalized name's components
                    parts = name.split('-')
                    if len(parts) >= 2 and parts[0] == 'llama':
                        # Handle LLaMA models
                        if len(parts) == 2:  # llama-2, llama-3
                            display_name = f"LLaMA-{parts[1].upper()}"
                        elif len(parts) == 3:  # llama-2-7b, llama-3-70b
                            display_name = f"LLaMA-{parts[1]}-{parts[2].upper()}"
                        elif len(parts) >= 4:  # llama-2-7b-chat, llama-3-8b-instruct
                            suffix = parts[3].capitalize() if len(parts) > 3 else ""
                            display_name = f"LLaMA-{parts[1]}-{parts[2].upper()}-{suffix}"
                    elif len(parts) >= 2 and parts[0] == 'mistral':
                        # Handle Mistral models
                        if len(parts) == 2:  # mistral-7b
                            display_name = f"Mistral-{parts[1].upper()}"
                        elif len(parts) >= 3:  # mistral-7b-instruct
                            suffix = parts[2].capitalize() if len(parts) > 2 else ""
                            display_name = f"Mistral-{parts[1].upper()}-{suffix}"
                    else:
                        # Use original name or the name as is
                        display_name = original_names.get(name, name.upper())
            else:  # benchmarks
                display_name = benchmark_display_names.get(name, original_names.get(name, name))
                
            ax.annotate(f"{display_name} ({counter[name]})",
                       xy=(date_max, baseline),
                       xytext=(5, 0),
                       textcoords="offset points",
                       va='center',
                       fontsize=9,
                       fontfamily='serif',
                       fontstyle='normal',
                       bbox=dict(facecolor='white',
                               edgecolor='none',
                               alpha=0.9,
                               pad=1))
    
    # Return the model baselines for adding release markers later
    return model_baselines

def create_combined_plot(df, names, counter, title, filename, color_h, original_names=None, plot_type=None):
    """Create a combined plot with paper counts and distribution"""
    plt.style.use('default')
    plt.rcParams.update({
        'font.family': 'serif',
        'font.size': 10,
        'axes.labelsize': 11,
        'axes.titlesize': 12,
        'figure.titlesize': 14
    })

    fig = plt.figure(figsize=(12, 10))
    
    # Create top subplot for paper counts (25% of height)
    ax1 = plt.subplot2grid((4, 1), (0, 0), rowspan=1)
    
    # Get total mentions per month
    monthly_mentions = df.groupby(df['date'].dt.to_period('M')).size()
    
    start_date = datetime(2022, 1, 1)
    end_date = datetime(2024, 12, 31)
    
    # First, filter papers_df to the date range we want
    papers_df_filtered = papers_df[
        (papers_df.index >= start_date) & 
        (papers_df.index <= end_date)
    ]
    
    # Create date range and align data
    monthly_dates = pd.date_range(start=start_date, end=end_date, freq='M')
    
    # Only include months where we actually have papers
    valid_months = papers_df_filtered.index.to_period('M')
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(valid_months)]
    
    # Resample papers data
    monthly_papers = papers_df_filtered.resample('M')['total_papers'].last()
    
    # Ensure both series align with actual paper months
    monthly_papers = monthly_papers[monthly_papers > 0]
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(monthly_papers.index.to_period('M'))]
    
    # Create date range for plotting
    start_date = datetime(2022, 1, 1)
    end_date = datetime(2024, 12, 31)
    monthly_dates = pd.date_range(start=start_date, end=end_date, freq='MS')  # MS = Month Start
    
    # Filter and align the data
    papers_df_filtered = papers_df[
        (papers_df.index >= start_date) & 
        (papers_df.index <= end_date)
    ]
    
    # Resample to month start frequency
    monthly_papers = papers_df_filtered.resample('MS')['total_papers'].last()
    monthly_mentions = df.groupby(df['date'].dt.to_period('M')).size()
    
    # Ensure both series align with actual paper months
    monthly_papers = monthly_papers[monthly_papers > 0]
    monthly_mentions = monthly_mentions[monthly_mentions.index.isin(monthly_papers.index.to_period('M'))]
    
    # Convert period index to timestamp for plotting
    plot_dates = pd.date_range(start=monthly_papers.index.min(), 
                             end=monthly_papers.index.max(), 
                             freq='MS')
    
    # Plot paper counts - align to start of each month
    bars = ax1.bar(plot_dates,
                   monthly_papers.values,
                   width=30,
                   color='gray',
                   alpha=0.5,
                   align='edge')  # Align to start of month

    # Customize top subplot
    ax1.set_xlim(start_date, end_date)
    ax1.spines['top'].set_visible(False)
    ax1.spines['left'].set_visible(False)
    ax1.spines['right'].set_visible(True)
    ax1.yaxis.set_label_position('right')
    ax1.yaxis.tick_right()
    
    # Add value labels for paper counts
    for bar in bars:
        height = bar.get_height()
        if height > 0:
            ax1.text(bar.get_x() + bar.get_width()/2, height,
                    f'{int(height)}',
                    ha='center', va='bottom', fontsize=8)
    
    # Create bottom subplot for distribution (75% of height)
    ax2 = plt.subplot2grid((4, 1), (1, 0), rowspan=3)
    
    n_shades = len(names)
    vertical_spacing = 0.35
    colors = sns.husl_palette(n_shades, h=color_h, s=0.85, l=0.6)
    
    # Plot ridgeline and get the model baselines
    model_baselines = plot_ridgeline(ax2, df, names, counter, colors, vertical_spacing, original_names, plot_type)
    
    # Add release date markers for each model if this is a models plot
    if plot_type == 'models':
        for name in names:
            # Check if this model has a release date
            if name in model_release_dates:
                release_date = datetime.strptime(model_release_dates[name], '%Y-%m-%d')
                
                # Check if the release date is within our plot range
                if start_date <= release_date <= end_date:
                    # Get the baseline y-coordinate for this model
                    baseline = model_baselines.get(name, 0)
                    
                    # Add a marker at the release date - just a simple tick mark
                    ax2.scatter([release_date], [baseline], 
                               marker='|',  # Vertical line marker
                               s=100,  # Size
                               color='black',  # Black color for visibility
                               linewidth=1.5,  # Thicker line
                               zorder=100)  # Ensure it's on top
    
    ax2.set_xlim(start_date, end_date)
    ax2.yaxis.set_visible(False)
    ax2.spines['top'].set_visible(False)
    ax2.spines['right'].set_visible(False)
    ax2.spines['left'].set_visible(False)
    
    # Format x-axis with year ticks
    for ax in [ax1, ax2]:
        # Set major ticks only at start of year
        ax.xaxis.set_major_locator(mdates.YearLocator(month=1, day=1))
        
        # Remove year labels from ticks
        ax.xaxis.set_major_formatter(plt.NullFormatter())
        
        # Handle year labels differently for each subplot
        if ax == ax1:  # Top subplot
            y_pos = ax.get_ylim()[0] - 0.1
        else:  # Bottom subplot (ax2)
            y_pos = -0.02  # Reduced negative value to move labels up
            
        # Manually add year labels at the middle of each year with larger, bolder font
        years = range(2022, 2025)
        for year in years:
            mid_year_date = datetime(year, 7, 1)
            ax.text(mdates.date2num(mid_year_date), y_pos,
                   str(year),
                   ha='center',
                   va='top',
                   transform=ax.get_xaxis_transform(),
                   fontsize=14,  # Increased font size
                   fontweight='bold')  # Made text bold
        
        # Add minor ticks for all months
        ax.xaxis.set_minor_locator(mdates.MonthLocator())
        
        # Customize tick lengths
        ax.tick_params(axis='x', which='major', length=8)  # Longer ticks for years
        ax.tick_params(axis='x', which='minor', length=4)  # Shorter ticks for months
        
        # Adjust label positions to center them
        ax.tick_params(axis='x', which='major', pad=15)  # Increased padding for year labels
        
        # Add subtle grid
        ax.grid(True, alpha=0.1, linestyle='-', axis='x')
        ax.grid(True, alpha=0.05, linestyle='-', axis='x', which='minor')
    
    # Only show x-label on bottom plot
    ax1.set_xlabel('')
    ax2.set_xlabel('')
    
    # Adjust layout
    plt.subplots_adjust(right=0.85, hspace=0.3)
    
    plt.savefig(filename,
                dpi=300,
                bbox_inches='tight',
                facecolor='white')
    plt.close()

def calculate_statistics(data, start_year=None):
    """Calculate statistics for the dataset with optional year filtering"""
    total_papers = len(set(title for title, paper_info in data.items() 
                          if 'publication_date' in paper_info 
                          and (start_year is None or 
                               datetime.strptime(paper_info['publication_date'], '%Y-%m-%d').year >= start_year)))

    total_model_mentions = 0
    total_benchmark_mentions = 0
    papers_with_models = 0
    papers_with_benchmarks = 0

    for title, paper_info in data.items():
        if 'publication_date' in paper_info and 'analysis' in paper_info:
            date = datetime.strptime(paper_info['publication_date'], '%Y-%m-%d')
            if start_year is None or date.year >= start_year:
                # Count model mentions
                if 'base_models' in paper_info['analysis'] and paper_info['analysis']['base_models']:
                    papers_with_models += 1
                    total_model_mentions += len(paper_info['analysis']['base_models'])
                
                # Count benchmark mentions
                if 'benchmarks' in paper_info['analysis'] and paper_info['analysis']['benchmarks']:
                    papers_with_benchmarks += 1
                    total_benchmark_mentions += len(paper_info['analysis']['benchmarks'])

    # Calculate averages
    avg_models_per_paper = total_model_mentions / total_papers if total_papers > 0 else 0
    avg_benchmarks_per_paper = total_benchmark_mentions / total_papers if total_papers > 0 else 0
    avg_models_when_present = total_model_mentions / papers_with_models if papers_with_models > 0 else 0
    avg_benchmarks_when_present = total_benchmark_mentions / papers_with_benchmarks if papers_with_benchmarks > 0 else 0

    return {
        'total_papers': total_papers,
        'total_model_mentions': total_model_mentions,
        'papers_with_models': papers_with_models,
        'avg_models_per_paper': avg_models_per_paper,
        'avg_models_when_present': avg_models_when_present,
        'total_benchmark_mentions': total_benchmark_mentions,
        'papers_with_benchmarks': papers_with_benchmarks,
        'avg_benchmarks_per_paper': avg_benchmarks_per_paper,
        'avg_benchmarks_when_present': avg_benchmarks_when_present
    }

# Main execution
if __name__ == "__main__":
    # Load and process data
    with open('papers_data_with_dates.json', 'r') as f:
        papers_data = json.load(f)

    # Calculate total papers per month
    all_papers_by_month = {}
    for title, paper_info in papers_data.items():
        if 'publication_date' in paper_info:
            date = paper_info['publication_date']
            date_obj = datetime.strptime(date, '%Y-%m-%d')
            month_key = date_obj.strftime('%Y-%m')
            all_papers_by_month[month_key] = all_papers_by_month.get(month_key, 0) + 1

    # Convert to DataFrame for plotting
    papers_df = pd.DataFrame(
        [(datetime.strptime(k, '%Y-%m'), v) for k, v in all_papers_by_month.items()],
        columns=['date', 'total_papers']
    )
    papers_df.set_index('date', inplace=True)

    # Get frequency counts
    model_counter = Counter()
    benchmark_counter = Counter()
    model_original_names = {}
    benchmark_original_names = {}
    benchmark_records = []
    model_records = []

    # Look for models that don't have a size (like standalone "LLaMA-2")
    standalone_models = set()

    # Process the data - FILTER OUT ANACHRONISTIC MENTIONS
    for title, paper_info in papers_data.items():
        if 'publication_date' in paper_info and 'analysis' in paper_info:
            date = paper_info['publication_date']
            date_obj = datetime.strptime(date, '%Y-%m-%d')
            
            # Only process data from 2022 onwards for charts
            if date_obj.year >= 2022:
                # Process benchmarks (no anachronism filtering for benchmarks)
                if 'benchmarks' in paper_info['analysis']:
                    for benchmark in paper_info['analysis']['benchmarks']:
                        normalized_benchmark = benchmark.strip('"').strip().lower()
                        normalized_benchmark = re.sub(r'[-_\s]+', '-', normalized_benchmark)
                        benchmark_counter[normalized_benchmark] += 1
                        benchmark_original_names[normalized_benchmark] = benchmark.strip('"').strip()
                        benchmark_records.append({
                            'date': date_obj,
                            'name': normalized_benchmark,
                            'type': 'Benchmark',
                            'paper': title
                        })
                
                # Process models - SKIP ANACHRONISTIC MENTIONS with enhanced checks
                if 'base_models' in paper_info['analysis']:
                    for model in paper_info['analysis']['base_models']:
                        normalized_model, original_name = normalize_model_name(model)
                        
                        # Enhanced anachronism check - more aggressive for LLaMA-3
                        if is_anachronistic_extra_check(normalized_model, date_obj):
                            continue
                        
                        # NEW: Apply consistent family normalization to strip chat/instruct variants
                        normalized_model = normalize_model_family(normalized_model)
                        
                        # Check for models without size specifications
                        if normalized_model in ['llama-2', 'llama-3'] and not re.search(r'\d+[bm]', normalized_model):
                            standalone_models.add((normalized_model, original_name))
                        
                        model_counter[normalized_model] += 1
                        if normalized_model not in model_original_names:
                            model_original_names[normalized_model] = original_name
                        model_records.append({
                            'date': date_obj,
                            'name': normalized_model,
                            'type': 'Model',
                            'paper': title
                        })

    # Apply more aggressive post-processing to fix any remaining LLaMA-3 issues
    model_records = post_process_model_records(model_records)
    
    # Rebuild the counter after post-processing
    model_counter = Counter([record['name'] for record in model_records])

    # Get top 20 benchmarks and models
    top_20_benchmarks = [b[0] for b in benchmark_counter.most_common(20)]
    top_20_models = [m[0] for m in model_counter.most_common(20)]

    # Create DataFrames
    df_benchmarks = pd.DataFrame(benchmark_records)
    df_models = pd.DataFrame(model_records)

    # Filter for top 20
    df_benchmarks = df_benchmarks[df_benchmarks['name'].isin(top_20_benchmarks)]
    df_models = df_models[df_models['name'].isin(top_20_models)]

    # Update DataFrames to use datetime
    df_benchmarks['date'] = pd.to_datetime(df_benchmarks['date'])
    df_models['date'] = pd.to_datetime(df_models['date'])

    # Calculate statistics for both full dataset and 2022-2024 period
    full_stats = calculate_statistics(papers_data)
    recent_stats = calculate_statistics(papers_data, 2022)

    # Print statistics for full dataset
    print("\nPaper Statistics (Full Dataset):")
    print(f"Total papers analyzed: {full_stats['total_papers']}")
    print(f"\nModel statistics:")
    print(f"Total model mentions: {full_stats['total_model_mentions']}")
    print(f"Papers that mention models: {full_stats['papers_with_models']} ({full_stats['papers_with_models']/full_stats['total_papers']*100:.1f}%)")
    print(f"Average models per paper (across all papers): {full_stats['avg_models_per_paper']:.2f}")
    print(f"Average models when present: {full_stats['avg_models_when_present']:.2f}")

    print(f"\nBenchmark statistics:")
    print(f"Total benchmark mentions: {full_stats['total_benchmark_mentions']}")
    print(f"Papers that mention benchmarks: {full_stats['papers_with_benchmarks']} ({full_stats['papers_with_benchmarks']/full_stats['total_papers']*100:.1f}%)")
    print(f"Average benchmarks per paper (across all papers): {full_stats['avg_benchmarks_per_paper']:.2f}")
    print(f"Average benchmarks when present: {full_stats['avg_benchmarks_when_present']:.2f}")

    # Print statistics for 2022-2024
    print("\nPaper Statistics (Jan 2022 - Dec 2024):")
    print(f"Total papers analyzed: {recent_stats['total_papers']}")
    print(f"\nModel statistics:")
    print(f"Total model mentions: {recent_stats['total_model_mentions']}")
    print(f"Papers that mention models: {recent_stats['papers_with_models']} ({recent_stats['papers_with_models']/recent_stats['total_papers']*100:.1f}%)")
    print(f"Average models per paper (across all papers): {recent_stats['avg_models_per_paper']:.2f}")
    print(f"Average models when present: {recent_stats['avg_models_when_present']:.2f}")

    print(f"\nBenchmark statistics:")
    print(f"Total benchmark mentions: {recent_stats['total_benchmark_mentions']}")
    print(f"Papers that mention benchmarks: {recent_stats['papers_with_benchmarks']} ({recent_stats['papers_with_benchmarks']/recent_stats['total_papers']*100:.1f}%)")
    print(f"Average benchmarks per paper (across all papers): {recent_stats['avg_benchmarks_per_paper']:.2f}")
    print(f"Average benchmarks when present: {recent_stats['avg_benchmarks_when_present']:.2f}")

    # Print top 20 benchmarks using original names
    print("\nTop 20 Benchmarks and their frequencies:")
    for benchmark, count in benchmark_counter.most_common(20):
        original_name = benchmark_original_names.get(benchmark, benchmark)
        display_name = benchmark_display_names.get(benchmark, original_name)
        print(f"{display_name}: {count}")

    # Print top 20 models using display names consistent with the chart
    print("\nTop 20 Models and their frequencies:")
    for model, count in model_counter.most_common(20):
        display_name = model_display_names.get(model, model_original_names.get(model, model))
        print(f"{display_name}: {count}")

    # Create combined plots
    create_combined_plot(df_benchmarks, top_20_benchmarks, benchmark_counter,
                        'Distribution of Benchmark Usage Over Time',
                        'benchmark_distribution_v2.pdf', 0.5,
                        benchmark_original_names,
                        'benchmarks')

    create_combined_plot(df_models, top_20_models, model_counter,
                        'Distribution of Model Usage Over Time',
                        'model_distribution_v2.pdf', 0.9,
                        model_original_names,
                        'models')
```
