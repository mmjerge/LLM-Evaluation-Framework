```python
import pandas as pd
import json

# Load the data from a JSON file
with open('/p/llmreliability/test_repos/ReAct/SoK_Experiments/results/SVAMP/gpt-3.5-turbo_svamp_react_results.json', 'r') as file:
    data = json.load(file)

# Convert the data to a pandas DataFrame
df = pd.DataFrame(data)

# Count the number of true and false values in the 'answered_correctly' column
result = df['answered_correctly'].value_counts()

true_count = result.get(True, 0)
false_count = result.get(False, 0)
total_count = true_count + false_count

print("Number of correct answers:", true_count)
print("Number of incorrect answers:", false_count)

# Calculate accuracy
if total_count > 0:
    accuracy = (true_count / total_count) * 100
    print(f"Accuracy: {accuracy:.2f}%")
else:
    print("No data available to calculate accuracy.")
```
