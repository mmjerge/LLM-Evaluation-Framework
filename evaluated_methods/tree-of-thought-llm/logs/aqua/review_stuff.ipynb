```python
import pandas as pd
import json

# Read the JSON file
with open('/p/llmreliability/test_repos/tree-of-thought-llm/logs/aqua/claude-3-5-sonnet-20240620_0.7_sample1_value1_greedy1_random150.json', 'r') as file:
    data = json.load(file)

# Convert the JSON data to a pandas DataFrame
df = pd.DataFrame(data)

# Create a new column 'is_correct' that compares extracted_answer to correct_answer
df['is_correct'] = df['extracted_answer'] == df['correct_answer']

# Count the number of correct responses
correct_count = df['is_correct'].sum()

# Calculate the total number of questions
total_questions = len(df)

# Calculate the accuracy
accuracy = correct_count / total_questions

print(f"Number of correct responses: {correct_count}")
print(f"Total number of questions: {total_questions}")
print(f"Accuracy: {accuracy:.2%}")

# Optional: Display a breakdown of correct/incorrect responses
print("\nBreakdown of responses:")
print(df['is_correct'].value_counts())

# Optional: List questions where the model response was incorrect
print("\nQuestions with incorrect responses:")
incorrect_df = df[~df['is_correct']]
for index, row in incorrect_df.iterrows():
    print(f"Dataset Index: {row['dataset_index']}")
    print(f"Question: {row['question']}")
    print(f"Model's extracted answer: {row['extracted_answer']}")
    print(f"Correct answer: {row['correct_answer']}")
    print()
```
