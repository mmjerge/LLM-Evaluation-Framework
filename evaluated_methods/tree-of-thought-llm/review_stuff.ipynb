```python
import pandas as pd
import json

# Read the JSON file
with open('/p/llmreliability/test_repos/tree-of-thought-llm/logs/mmlu/gpt-3.5-turbo_0.7_propose1_value1_greedy1_start0_end100.json', 'r') as file:
    data = json.load(file)

# Create a DataFrame
df = pd.DataFrame(data)

# Calculate accuracy
df['is_correct'] = df['model_response'] == df['correct_answer']
accuracy = df['is_correct'].mean()

# Print results
print(f"Total questions: {len(df)}")
print(f"Correct answers: {df['is_correct'].sum()}")
print(f"Accuracy: {accuracy:.2%}")
```
