```python
import pandas as pd

# Read the data
df = pd.read_json('open-mixtral-8x22b_gsm8k_results.json')

# Improved regex patterns that handle more cases
ground_truth_pattern = r'####\s*[$₺£€¥₣]?\s*(\d+(?:\.\d+)?)'
model_answer_pattern = r'Final answer:.*?(?:####)?\s*[$₺£€¥₣]?\s*\$?\\?\$?\s*(\d+(?:\.\d+)?)'

# Extract answers
df['ground_truth'] = df['correct_answer'].str.extract(ground_truth_pattern).astype(float)
df['model_answer'] = df['model_response'].str.extract(model_answer_pattern).astype(float)

# Calculate accuracy
df['correct'] = df['ground_truth'] == df['model_answer']
results = df[['ground_truth', 'model_answer', 'correct']]
print(results)

# Print accuracy statistics
accuracy = (df['correct'].sum() / len(df)) * 100
print(f"\nAccuracy: {accuracy:.2f}%")
```
